```python
class Person:
    def __init__(self, name):
        self.__name = name  # Protected variable

    @property
    def name(self):
        """Getter method that acts like an attribute."""
        return self.__name
    
    @name.setter
    def set_name(self,new_name):
        self.__name = new_name
        print('New Name has been set')

p = Person("Alice")
```

```python
class Animal:
    def __init__(self, name):
        self.name = name  # `self` still refers to subclass instance

    def parent_method(self):
        print('Hello I am parent method')

    def speak(self, speak="Bark"):
        print(f"The dog: {speak}")

class Dog(Animal):
    def __init__(self, name, breed):
        super().__init__(name)  # Calls `Animal.__init__`
        self.breed = breed
    
    def speak(self, speak=None):
        if speak is not None:  # Fixed condition
            print(f'The dog says: {speak}')
        else:
            super().speak()  # Pass `speak` to parent method

# Creating an object
dog = Dog("Max", "Labrador")

print(dog.name)   # Output: Max
print(dog.breed)  # Output: Labrador

dog.parent_method()  # Output: Hello I am parent method
dog.speak()  # Output: The dog says: Bark (default value)
dog.speak("Woof!")  # Output: The dog says: Woof!
```

```python
import time
import threading

shared_list = []
lock = threading.Lock()

def print_numbers():
    start_time = time.time()
    for num in range(5):
        with lock:
            shared_list.append(num)
        print(f"{time.time():.5f} Number : {num}")
        # print()
        time.sleep(1)
    end_time = time.time()
    print(f'I am T1 total time : {end_time-start_time:.5f}')


def print_alpha():
    start_time = time.time()
    for char in "ABCDE":
        with lock:
            shared_list.append(char)
        print(f"{time.time():.5f} Alphabet : {char.lower()}")
        # print()
        time.sleep(1)
    end_time = time.time()
    print(f'I am T2 total time : {end_time-start_time:.5f}')


t1 = threading.Thread(target=print_numbers)
t2 = threading.Thread(target=print_alpha)
# threading.Thread()
total_start_time = time.time()
t1.start()
t2.start()

t1.join()
t2.join()

print(f"total time is {time.time()-total_start_time:.5f}")
print('Threads execution done!')
```

```python
def fibo_series(n):
    if n<=1:
        return n
    else:
        return(fibo_series(n-1) + fibo_series(n-2))

for i in range(10):
    print(fibo_series(i))
```

```python
def is_valid_parentheses(s):
    stack = []
    mapping = {')': '(', '}': '{', ']': '['}  # Closing → Opening mapping

    for char in s:
        if char in mapping:  # If it's a closing bracket
            top_element = stack.pop() if stack else '#'  # Pop last open bracket or use placeholder
            if mapping[char] != top_element:  # Check if it matches expected opening bracket
                return False
        else:  # If it's an opening bracket
            stack.append(char)

    return not stack  # Stack should be empty if valid

# ✅ Test Cases
print(is_valid_parentheses("()[]{}"))  # ✅ True
print(is_valid_parentheses("([{}])"))  # ✅ True
print(is_valid_parentheses("(]"))      # ❌ False
print(is_valid_parentheses("([)]"))    # ❌ False
print(is_valid_parentheses("{[]}"))    # ✅ True
```

```python
from multiprocessing import Process, Queue
import time

def producer(q):
    for i in range(5):
        q.put(i)  # Put data into the queue
        print(f"Produced: {i}")
        time.sleep(1)

def consumer(q):
    while not q.empty():
        item = q.get()  # Retrieve data from the queue
        print(f"Consumed: {item}")
        time.sleep(1)

if __name__ == "__main__":
    q = Queue()
    p1 = Process(target=producer, args=(q,))
    p2 = Process(target=consumer, args=(q,))
    
    p1.start()
    p1.join()  # Ensure producer finishes before consumer starts

    p2.start()
    p2.join()
```
