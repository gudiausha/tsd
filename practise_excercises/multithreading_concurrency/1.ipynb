Python program to create multiple threads and print their names

```python
import threading

def print_thread_names():
    print(f'Printing from {threading.current_thread().name} and id is {threading.current_thread().ident}')

threads_list = []

for i in range(5):
    thread = threading.Thread(target=print_thread_names)
    threads_list.append(thread)
    thread.start()

for thread_obj in threads_list:
    thread.join()
```

Python script to spawn a user-specified number of threads, have each thread sleep for a random time, and print its name upon waking up

```python
import threading
import time
import random

class ThreadsScript:
    def __init__(self,n):
        self.n = n

    def task(self):
        start_time = time.time()
        print(f'Creating {threading.current_thread().name}')
        print(f'{threading.current_thread().name} sleeping')
        time.sleep(random.randint(1,5))
        print(f'Hi I am thread - {threading.current_thread().name}')
        print(f'{threading.current_thread().name} execution time is {time.time()-start_time:.5f}')

    def main(self):
        threads_list = []
        program_start = time.time()
        for i in range(self.n):
            thread = threading.Thread(target=self.task)
            threads_list.append(thread)
            thread.start()

        for thread_obj in threads_list:
            thread_obj.join()

        print(f'Main program execution time - {time.time()-program_start:.5f}')

obj1 = ThreadsScript(3)
obj1.main()
```
