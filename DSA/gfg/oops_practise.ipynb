Encapsulation

```python
class Person:
    def __init__(self):
        self.__name = "Geeks"
        self.__age = 10

    def get_name(self):
        print(self.__name)

    def get_age(self):
        print(self.__age)

    def set_name(self,name_inp):
        self.__name = name_inp

    def set_age(self,age_inp):
        self.__age = age_inp

obj_person = Person()
obj_person.get_name()
obj_person.get_age()
obj_person.set_age(20)
obj_person.get_age()
obj_person.set_name('ABCD')
obj_person.get_name()
```

```python
class Person:
    def __init__(self):
        self.__name = "Geeks"
        self.__age = 10

    @property
    def get_name(self):
        print(self.__name)

    @property
    def get_age(self):
        print(self.__age)

    @get_name.setter
    def set_name(self,name_inp):
        self.__name = name_inp

    @get_age.setter
    def set_age(self,age_inp):
        self.__age = age_inp


obj = Person()
obj.get_age
```

Abstraction

```python
from abc import ABC, abstractmethod

# Abstract class Shape
class Shape(ABC):
    def __init__(self, c: str):
        self.color = c

    def get_color(self) -> str:
        return self.color

    @abstractmethod
    def get_area(self) -> float:
        pass  # Abstract method

# Concrete class Square extending Shape
class Square(Shape):
    def __init__(self, c: str, side: float):
        super().__init__(c)  # Initialize color using the parent class
        self.side = side

    def get_area(self) -> float:
        return self.side * self.side

# Driver code
if __name__ == "__main__":
    square = Square("Red", 5.0)  # Create a Square object
    print("Color:", square.get_color())  # Output: Red
    print("Area:", square.get_area())  # Output: 25.0
```
