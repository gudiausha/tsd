A general list of common time complexities arranged in **ascending order** (from fastest to slowest):

| **Notation**           | **Name**                           | **Example Use Cases**                                           |
|------------------------|------------------------------------|----------------------------------------------------------------|
| **O(1)**                | Constant Time                     | Accessing array element, hash map lookup                        |
| **O(log log N)**        | Double Logarithmic Time           | Some advanced algorithms (rare)                                 |
| **O(log N)**            | Logarithmic Time                  | Binary Search, Balanced BST operations (insert/search/delete)   |
| **O(√N)**               | Square Root Time                  | Primality check (naive), some divide & conquer algorithms       |
| **O(N)**                | Linear Time                       | Traversing arrays, linked lists                                 |
| **O(N log N)**          | Linearithmic Time                 | Efficient sorting (Merge Sort, Heap Sort, QuickSort avg case)   |
| **O(N²)**               | Quadratic Time                    | Bubble Sort, Selection Sort, Insertion Sort, naive DP solutions |
| **O(N² log N)**         | Slightly higher than Quadratic    | Some graph algorithms                                           |
| **O(N³)**               | Cubic Time                        | Floyd-Warshall algorithm, Triple nested loops                   |
| **O(2ᴺ)**               | Exponential Time                  | Recursive solutions without memoization (e.g., Fibonacci)       |
| **O(N!)**               | Factorial Time                    | Brute-force permutations, Traveling Salesman Problem            |

---

### **Quick Reference (Just the Orders)**:

```
O(1) < O(log log N) < O(log N) < O(√N) < O(N) < O(N log N) < O(N²) < O(N² log N) < O(N³) < O(2ᴺ) < O(N!)
```

---

**Note:**
- Lower complexity means faster algorithms.
- Constants and lower-order terms are usually ignored in Big-O, but they do matter in practice.
- Exponential and factorial complexities are generally impractical for large inputs.


## Contains Duplicate

````python
def containsDuplicate(nums) -> bool:
    hash_set=set()  #Extra Space - O(N)
    for ele in nums: #O(N)
        if ele in hash_set: #O(1)
            return True
        hash_set.add(ele)  #O(1)
    return False
````

### **Step-by-Step Complexity Breakdown:**
1. **Looping through `nums` (O(N))**  
   - The function iterates over all elements in `nums`, making it **O(N)**.
   
2. **Checking if an element exists in `hash_set` (`if ele in hash_set`)**  
   - Set lookup operations (`in` operator) have **O(1) average time complexity**.

3. **Inserting into `hash_set` (`hash_set.add(ele)`)**  
   - Inserting an element into a set is **O(1) average time complexity**.

4. **Worst Case Scenario (No Duplicates Exist)**  
   - The function goes through **all `N` elements** without returning early.
   - **Overall TC: O(N)**.

5. **Space Complexity (`O(N)`)**
   - In the worst case (all elements unique), we store **all `N` elements** in `hash_set`.
   - **SC = O(N)**.

---

### **Optimal Approach**
The given approach using a **set** is already optimal in terms of **time complexity (O(N))**. However, if modifying the input is allowed, we can **sort the array** and check adjacent elements:

✅ **Time Complexity: O(N log N) (due to sorting)**  
✅ **Space Complexity: O(1) (if in-place sorting is used)**  

#### **Comparison:**
| Approach         | Time Complexity | Space Complexity |
|-----------------|---------------|-----------------|
| **Hash Set**    | **O(N)**       | **O(N)**       |
| **Sorting**     | **O(N log N)**  | **O(1) (in-place)** |

### **Conclusion**
- If we want the **best time complexity**, **hash set approach (O(N))** is optimal.
- If we **must use O(1) space**, the **sorting approach (O(N log N))** is better.

## Valid Anagram


````python
def isAnagram(s: str, t: str) -> bool:
    from collections import Counter
    s_dict = Counter(s)
    t_dict = Counter(t)
    return s_dict==t_dict
````

### **Time Complexity (TC) & Space Complexity (SC) Analysis**
### **Step-by-Step Analysis:**
1. **Counter(s) and Counter(t)**  
   - `Counter(s)`: Creates a frequency dictionary of characters in `s`.
   - `Counter(t)`: Creates a frequency dictionary of characters in `t`.
   - `Counter()` iterates through the string **once** and counts the frequency of each character.  
   - **Time Complexity**: \(O(N)\), where \(N\) is the length of the string.
   - **Space Complexity**: \(O(1)\), since there are at most **26 lowercase English letters** (constant space).

2. **Comparison (`s_dict == t_dict`)**  
   - Since both `Counter` dictionaries store at most **26 unique keys**, the comparison is **constant time** \(O(1)\).
   - **Time Complexity**: \(O(1)\) in the worst case.

### **Final Complexity:**
- **Time Complexity**: \(O(N)\)  
- **Space Complexity**: \(O(1)\) *(constant space since only 26 letters are stored)*  

---

### **Alternative Approach Using Sorting**
- **Sorting takes** \(O(N \log N)\)
- **Space Complexity**: \(O(N)\) due to sorting storage.

Thus, the **Counter method is optimal** compared to sorting. 🚀

## Two Sum - input arr not sorted


````python
def twoSum(nums,target):
    hash_set = {}
    for ind,num in enumerate(nums):
        complement = target-num
        if complement in hash_set:
            return [hash_set[complement],ind]
        hash_set[num] = ind
    return [0,0]
````


### **Time Complexity (TC) & Space Complexity (SC) Analysis**
---
### **Step-by-Step Complexity Breakdown:**
1. **Looping through `nums` (O(N))**
   - The function iterates through all elements of `nums`, making it **O(N)**.

2. **Checking if `complement` exists in `hash_set` (`if complement in hash_set`)**
   - Dictionary lookup takes **O(1) average time**.
   - This operation is performed once per iteration.

3. **Inserting into `hash_set` (`hash_set[num] = ind`)**
   - Dictionary insertion takes **O(1) average time**.

4. **Early Exit When a Solution is Found**
   - The loop exits immediately once a solution is found.
   - In the worst case (no duplicate complements), we iterate through **all `N` elements**.
   - **Overall TC: O(N)**.

---

### **Space Complexity (`O(N)`)**
- In the worst case (when no solution is found early), we store **all `N` elements** in `hash_set`.
- Since we store key-value pairs in the dictionary, the space required is **O(N)**.

---

### **Comparison with Brute Force Approach**
| Approach         | Time Complexity | Space Complexity |
|-----------------|---------------|-----------------|
| **Brute Force (Nested Loops)**  | **O(N²)**  | **O(1)** |
| **Hash Map (Optimal)** | **O(N)** | **O(N)** |

---

### **Conclusion**
🚀 The **hash map approach (O(N) TC, O(N) SC)** is the optimal way to solve this problem efficiently.

## Group Anagrams

````python
def groupAnagrams(strs):
    hash_set = {}
    for word in strs:
        char_arr = [0]*26
        for letter in word:
            char_arr[ord(letter)-ord('a')]+=1
        key = tuple(char_arr)
        if key in hash_set:
            hash_set[key].append(word)
        else:
            hash_set[key] = [word]

    return list(hash_set.values())
````

### **Time Complexity (TC) & Space Complexity (SC) Analysis**
---

### **Time Complexity (O(N * K))**
#### **Breakdown**
1. **Iterating through `strs` (O(N))**  
   - We iterate over `N` words.

2. **Processing Each Word (O(K))**  
   - We iterate over each word (max length `K`) to count character occurrences (`O(K)`).
   - We then create a tuple from the 26-length list (`O(1)`, since it's fixed-size).

3. **Dictionary Operations**
   - **Checking for key existence (`if key in hash_set`)**: Average **O(1)**.
   - **Appending to list (`hash_set[key].append(word)`)**: Average **O(1)**.
   - **Inserting a new key (`hash_set[key] = [word]`)**: Average **O(1)**.

Since each word requires **O(K)** operations, and we process **N** words, the total complexity is:
\[
O(N * K)
\]

---

### **Space Complexity (O(N * K))**
1. **Dictionary (`hash_set`) Storage: O(N * K)**  
   - The keys (tuples of length 26) are fixed-size **O(1)**.
   - The values (lists of words) store all input words (`O(N * K)` in worst case).

2. **Character Frequency Array (`char_arr`): O(1)**  
   - This is a fixed-size **26-element array**, so it takes **constant space O(1)** per word.

Thus, the dominant term is **O(N * K)**.

---

### **Comparison with Sorting-Based Approach**
| Approach | Time Complexity | Space Complexity |
|----------|---------------|-----------------|
| **Sorting Each Word (`sorted(word)`)** | **O(N * K log K)** | **O(N * K)** |
| **Character Frequency Tuple (Optimal)** | **O(N * K)** | **O(N * K)** |

**🚀 Conclusion:** The frequency count approach is more efficient than sorting-based grouping because **sorting takes O(K log K) per word, while frequency counting is O(K)**.

## Top k frequent elements

````python
def topKFrequent(nums, k):
    count = {}
    freq = [[] for i in range(len(nums)+1)]

    for num in nums:
        count[num] = 1 + count.get(num,0)

    for key,value in count.items():
        freq[value].append(key)

    result = []
    for i in range(len(freq)-1,0,-1):
        for num in freq[i]:
            result.append(num)
            if len(result)==k:
                return result
````

### **Step-by-Step Explanation & Dry Run**
The function **`topKFrequent(nums, k)`** finds the **top K most frequent elements** in a list of integers using **bucket sorting**. 

---

### **Algorithm Breakdown**
1. **Count the Frequency of Each Number**  
   - We use a dictionary `count` to store the frequency of each number in `nums`.  
   
2. **Use Bucket Sort (List of Lists `freq`)**  
   - `freq[i]` will store numbers that appear exactly `i` times.
   - We initialize `freq` as a list of empty lists, with size `len(nums) + 1` (maximum frequency possible is `len(nums)`).
  
3. **Populate `freq` Buckets**  
   - Traverse `count` and insert each number into `freq[freq_count]`, where `freq_count` is the frequency of that number.

4. **Retrieve Top K Frequent Elements**  
   - Iterate **backward** through `freq` (starting from the most frequent elements).  
   - Append elements to `res` until we collect `k` elements.

---

### **Dry Run Example**
### **Input:**
```python
nums = [1,1,1,2,2,3]
k = 2
```
---
### **Step 1: Frequency Count**
We create a **dictionary `count`** to store the number of occurrences for each element:

| Number | Frequency |
|--------|----------|
| 1      | 3        |
| 2      | 2        |
| 3      | 1        |

```python
count = {1: 3, 2: 2, 3: 1}
```

---
### **Step 2: Bucket Sort (Store Elements in `freq`)**
- Create an **array of lists** (`freq`) with size `len(nums) + 1 = 7`:
```python
freq = [[], [], [], [], [], [], []]
```
- Traverse `count` and place each number in `freq[freq_count]`:

| `freq` Index (i) | Numbers Appearing `i` Times |
|-----------------|---------------------------|
| 0               | []                        |
| 1               | [3]                       |
| 2               | [2]                       |
| 3               | [1]                       |
| 4               | []                        |
| 5               | []                        |
| 6               | []                        |

```python
freq = [[], [3], [2], [1], [], [], []]
```

---
### **Step 3: Extract Top K Elements**
- Start **backward** from `freq[len(nums)]` (index `6` to `0`).
- Find the **top `k=2` most frequent elements**.

| Iteration (`i`) | Elements in `freq[i]` | `res` (Result List) |
|----------------|-----------------|-----------------|
| `i=6`         | `[]`             | `[]`           |
| `i=5`         | `[]`             | `[]`           |
| `i=4`         | `[]`             | `[]`           |
| `i=3`         | `[1]`            | `[1]`          |
| `i=2`         | `[2]`            | `[1, 2]`       |

Since we have found `k=2` elements, **return `[1, 2]`**.

---
### **Final Output**
```python
[1, 2]
```
---
### **Time Complexity Analysis**
1. **Counting Frequency (`count[num] = 1 + count.get(num, 0)`)** → **O(N)**
2. **Storing in Buckets (`freq[cnt].append(num)`)** → **O(N)**
3. **Collecting Top K (`for i in range(len(freq) - 1, 0, -1)`)** → **O(N)** (worst case when all elements have unique frequencies)
4. **Total Complexity** → **O(N)** (since each step is linear)

---
### **Space Complexity Analysis**
1. **`count` dictionary** → Stores at most `N` elements → **O(N)**
2. **`freq` array of lists** → Stores at most `N+1` elements → **O(N)**
3. **`res` list** → Stores `K` elements → **O(K)**
4. **Total Space Complexity** → **O(N)**

---
### **Key Takeaways**
- **Bucket Sort** ensures `O(N)` time complexity.
- **Dictionary (`Counter`) + List (`freq`)** make this approach very efficient.
- Works better than sorting approaches (`O(N log N)`).

## Encode and Decode Strings

````python
def encode(str_list):
    limit_char = '$'
    encoded_str = ''

    for word in str_list:
        word_len = str(len(word))
        encoded_str += word_len+limit_char+word
    
    return encoded_str

def decode(encoded_str):
    decoded_output = []
    i = 0
    while i < len(encoded_str):
        j = i
        while encoded_str[j] != '$':
            j+=1
        lenght = int(encoded_str[i:j])
        i = j+1
        j = i + lenght
        decoded_output.append(encoded_str[i:j])
        i = j
    return decoded_output
    
````

### **Explanation**
### **Encoding Process**
The `encode` function **converts a list of strings** into a single string using a delimiter (`$`):
- **Each word is prefixed with its length**, followed by `$`, and then the word itself.

🔹 **Example:**  
```python
words = ["hello", "world"]
encoded = encode(words)
print(encoded)  # "5$hello5$world"
```
- `"hello"` → `5$hello`
- `"world"` → `5$world`
- Final result: `"5$hello5$world"`

---

### **Decoding Process**
The `decode` function **extracts words** from the encoded string:
1. **Extract the length** of each word (before `$`).
2. **Read the next 'length' characters** to get the word.
3. **Repeat the process** until all words are extracted.

🔹 **Example:**
```python
encoded = "5$hello5$world"
decoded = decode(encoded)
print(decoded)  # ["hello", "world"]
```
- **First word:**
  - Read `5`, then extract **5 characters** → `"hello"`
- **Second word:**
  - Read `5`, then extract **5 characters** → `"world"`


---

### **Time & Space Complexity**
### **Encoding:**
- **Time Complexity:** \(O(N)\), where \(N\) is the total number of characters in `str_list`.
- **Space Complexity:** \(O(N)\), storing the encoded string.

### **Decoding:**
- **Time Complexity:** \(O(N)\), since we traverse the encoded string once.
- **Space Complexity:** \(O(N)\), storing the decoded list.

## Product of Array except self

````python
def ProductExceptSelf(arr):
    n = len(arr)
    ans = [1]*n

    l=1
    for i in range(n):
        ans[i]=l
        l*=arr[i]

    r = 1
    for i in range(n-1,-1,-1):
        ans[i]*=r
        r *= arr[i]

    return ans
````

### **Step-by-Step Explanation**
This function **computes the product of all elements in an array except the current element** without using division.

---

### **Idea: Left and Right Products**
- We **cannot** directly divide the total product by the current element (as it might contain zeroes).  
- Instead, we use two passes:
  1. **Left product pass**: Compute the product of elements **before** the current index.
  2. **Right product pass**: Compute the product of elements **after** the current index.
  3. **Multiply both left and right products** to get the final answer.

---

### **Dry Run**
#### **Input:**  
```python
arr = [1, 2, 3, 4]
```
#### **Step 1: Initialize Answer Array**
We create an `ans` array initialized to `[1, 1, 1, 1]`.

---

### **Left Product Pass**
We compute products **before** each element.

| `i`  | `l` (before update) | `ans[i] = l` | `l *= arr[i]` (update `l`) |
|------|------------------|------------|------------------|
| 0    | 1                | 1          | `l = 1 * 1 = 1`  |
| 1    | 1                | 1          | `l = 1 * 2 = 2`  |
| 2    | 2                | 2          | `l = 2 * 3 = 6`  |
| 3    | 6                | 6          | `l = 6 * 4 = 24` |

So, `ans` becomes:
```python
ans = [1, 1, 2, 6]
```

---

### **Right Product Pass**
We compute products **after** each element.

| `i`  | `r` (before update) | `ans[i] *= r` | `r *= arr[i]` (update `r`) |
|------|------------------|--------------|------------------|
| 3    | 1                | `6 * 1 = 6`   | `r = 1 * 4 = 4`  |
| 2    | 4                | `2 * 4 = 8`   | `r = 4 * 3 = 12` |
| 1    | 12               | `1 * 12 = 12` | `r = 12 * 2 = 24`|
| 0    | 24               | `1 * 24 = 24` | `r = 24 * 1 = 24`|

Now `ans` becomes:
```python
ans = [24, 12, 8, 6]
```

---

### **Final Output**
```python
ProductExceptSelf([1, 2, 3, 4])  # Output: [24, 12, 8, 6]
```

---

### **Time and Space Complexity**
- **Time Complexity:** \(O(n)\) (Two passes, each taking \(O(n)\))
- **Space Complexity:** \(O(1)\) (Excluding output array, since it doesn't count towards extra space in complexity analysis)

---

### **Summary**
✅ **No division used**  
✅ **Only two passes (O(n) time)**  
✅ **Uses only constant extra space**  


## Best time to buy and sell stock

````python
#Using DP
def buyandsellstock(prices):
    maxprofit = 0
    minbuy = prices[0]

    for sell in prices:
        current_profit = sell-minbuy
        maxprofit = max(maxprofit,current_profit)
        minbuy = min(minbuy,sell)

    return maxprofit
````

### **Time Complexity (TC)**
- The loop iterates over the `prices` array **once** → **O(N)**
- Inside the loop, all operations (subtraction, `max()`, and `min()`) take **O(1)**.
- Thus, **TC = O(N)**.

### **Space Complexity (SC)**
- Only two extra variables (`maxp`, `minbuy`) are used → **O(1)** auxiliary space.
- Thus, **SC = O(1)**.

---

````python
#Using 2 pointer
def buyandsellstock(prices):
    l = 0
    r = 1
    maxprofit = 0
    while r<len(prices):
        if prices[l] < prices[r]:
            currentprofit = prices[r]-prices[l]
            maxprofit = max(maxprofit,currentprofit)
        else:
            l=r
        
        r+=1
    return maxprofit

    
````

## Valid Parentheses

````python
def isValid(s):
    stack = []
    closeToopen = {')':'(','}':'{',']':'['}

    for char in s:
        if char in closeToopen:
            top_ele = stack.pop() if stack else '#'
        
            if closeToopen[char] != top_ele:
                return False
        
        else:
            stack.append(char)
    
    return not stack
````

### **Time Complexity (TC)**
- We iterate over the string **once** → **O(N)**.
- Stack operations (`push`, `pop`) are **O(1)**.
- Thus, **TC = O(N)**.

### **Space Complexity (SC)**
- In the worst case (e.g., `"(((([[[{{{"`), the stack holds all characters → **O(N)**.
- Thus, **SC = O(N)**.

---

### **Dry Run**
Let's dry run the algorithm with an example.

### **Example Input**
```python
s = "{[()]}"
```

### **Initial State**
| Stack  | Current Char | Action |
|--------|-------------|--------|
| `[]`   | `{`         | Push `{` |
| `[{]`  | `[`         | Push `[` |
| `[{[`  | `(`         | Push `(` |
| `[{[`  | `)`         | Pop `(` (matches) |
| `[{]`  | `]`         | Pop `[` (matches) |
| `[{}`  | `}`         | Pop `{` (matches) |

### **Final Output**
- Stack is empty → **Valid string** → Returns `True`.

---

### **Edge Cases Handled**
1. **Empty String (`s = ""`)**
   - Stack remains empty → Valid → Returns `True`.
2. **Unbalanced Cases**
   - `"(]"` → `False`
   - `"{[}"` → `False`
   - `"((((("` → `False`
3. **Extra Closing Brackets**
   - `"))"` → `False`
4. **Extra Opening Brackets**
   - `"((("` → `False`
5. **Mixed Correctly Ordered Brackets**
   - `"({[]})"` → `True`
6. **Large Input (`s` with millions of brackets)**
   - **O(N) efficiency** makes it **scalable**.


## Valid Palindrome

````python
def isPalindrome(s: str) -> bool:
    cleaned_s = ''
    for char in s:
        if char.isalnum():
            cleaned_s+=char.lower()

    l = 0
    r = len(cleaned_s)-1
    while l <=r:
        if cleaned_s[l] == cleaned_s[r]:
            l+=1
            r-=1
        else:
            return False
    return True
````

### **Understanding the Code: `isPalindrome`**
This function checks whether a given string `s` is a **palindrome** after:
1. Removing **non-alphanumeric characters** (letters and digits are kept, symbols are ignored).
2. Converting all letters to **lowercase** (to make it case-insensitive).
3. Using **two-pointer technique** to compare characters from both ends.

---

### **Step-by-Step Explanation**
### **1. Cleaning the Input String**
```python

for char in s:
    if char.isalnum():
        cleaned_s += char.lower()
```
- `isalnum()` ensures that only **letters (a-z, A-Z) and digits (0-9)** are kept.
- `lower()` ensures that **uppercase and lowercase letters** are treated equally.
- The cleaned string is built by iterating through `s` character by character.

### **2. Checking for Palindrome Using Two Pointers**
```python
l = 0
r = len(cleaned_s) - 1
while l <= r:
    if cleaned_s[l] == cleaned_s[r]:
        l += 1
        r -= 1
    else:
        return False
return True
```
- **Left (`l`) starts at 0** and **Right (`r`) starts at the last character**.
- If `cleaned_s[l]` matches `cleaned_s[r]`, move both pointers inward (`l+1`, `r-1`).
- If any mismatch occurs, **return `False`** immediately.
- If the loop completes, the string is a palindrome → **return `True`**.

## Two Sum II - Input arr sorted

## Binary Search

````python

def search(nums,target) -> int:
    l = 0
    r = len(nums) - 1

    while l <= r:
        mid = l + (r - l) // 2  # Avoids overflow in some languages
        
        if nums[mid] == target:
            return mid  # Found target
        
        elif nums[mid] < target:
            l = mid + 1  # Search right half
        
        else:
            r = mid - 1  # Search left half
    
    return -1  # Target not found
````

### **Time Complexity (TC)**
- Each iteration halves the search space.
- In the worst case, the number of iterations is **log₂(N)**.
- Thus, **TC = O(log N)**.

### **Space Complexity (SC)**
- **O(1)**, as we use only a few extra variables (`l`, `r`, `mid`).

---

## Longest Consecutive Number

````python
def longest_consecutive_num(num):
    if not num:
        return 0
    
    nums_set = set(num)
    longest = 0

    for number in nums_set:
        if number-1 not in nums_set:
            streak=1
            current_num=num
        
        while current_num+1 in nums_set:
            current_num+=1
            streak+=1
        
        longest = max(longest,streak)
    
    return longest
````

### **Time Complexity (TC)**
- **O(N)** because:
  - **Creating the set** takes **O(N)**.
  - **Iterating through the set** takes **O(N)**.
  - **Checking `num - 1 in num_set`** is **O(1)** per element.
  - **Checking `num + 1 in num_set` repeatedly** runs **in total O(N)** across all elements.

### **Space Complexity (SC)**
- **O(N)** because:
  - The `num_set` stores at most **N** unique numbers.
  - A few extra variables are used (`longest`, `current_num`, `current_streak`).

## 3 sum

````python
def three_sum(nums):
    res = []
    nums.sort()  # Sort the array

    for i in range(len(nums)):
        if i > 0 and nums[i] == nums[i - 1]:  # Skip duplicates
            continue

        j = i + 1
        k = len(nums) - 1

        while j < k:
            total = nums[i] + nums[j] + nums[k]

            if total > 0:
                k -= 1
            elif total < 0:
                j += 1
            else:
                res.append([nums[i], nums[j], nums[k]])
                
                # Move `j` and `k` to avoid duplicates
                j += 1
                k -= 1
                
                # Skip duplicate elements for `j`
                while j < k and nums[j] == nums[j - 1]:  
                    j += 1

                # Skip duplicate elements for `k`
                while j < k and nums[k] == nums[k + 1]:
                    k -= 1

    return res
````

### **Time Complexity (TC)**
1. **Sorting `nums` takes O(N log N)**.
2. **Outer loop runs O(N) times**.
3. **Inner two-pointer loop runs in O(N)** in total across all `i`.
4. **Overall Complexity: O(N log N) + O(N²) ≈ O(N²).**

### **Space Complexity (SC)**
- **O(1)** (ignoring output list), since we modify `nums` in place and only use pointers (`i`, `j`, `k`).
- **O(M)** for storing results (`M` is the number of unique triplets).
