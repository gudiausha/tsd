## 2 sum problem
- Given an array of integers nums and an integer target, return indices of the two numbers such that they add up to target.
- *Assumption*: Exactly one solution is possible
- *Thought process*:
    - Is the arr sorted (2 pointer works fine)
    - Unsorted (Dictionary works fine)

```python
def two_sum_sorted(arr,target):
    i = 0
    j = len(arr)-1
    while i<j:
        if arr[i]+arr[j] == target:
            return i,j       
        elif arr[i]+arr[j] < target:
            i+=1
        else:
            j-=1
    return -1

output = two_sum_sorted([2,7,11,15],9)
```

```python
output
```

```python
def hash_set_unsorted(arr,target):
    complement_set = set()
    for num in arr:
        complement = target-num
        if complement in complement_set:
            return num,complement
        else:
            complement_set.add(num)
    return -1

output = hash_set_unsorted([2,5,3,4,9],6)
output
```
