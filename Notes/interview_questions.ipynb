### Q1: Explain Dynamic Typing , Type Inference , `is` vs `==` and Type Casting

#### 1. 🌀 **Dynamic Typing**

#### 📌 Definition:
Python is **dynamically typed**, meaning you **don’t need to declare variable types** — the type is inferred at runtime.

#### ✅ Example:
```python
x = 10       # int
x = "Hello"  # now a str
```

You can reassign `x` from an int to a str, and Python doesn't complain.

#### 🔍 Interview Insight:
> "Dynamic typing allows flexibility but can lead to runtime errors if types are misused. Static type checks (via tools like mypy) can be added for safety."

#### 2. 🧠 **Type Inference**

#### 📌 Definition:
Type inference is when the language **guesses the variable type based on the value assigned**.

In Python, this happens at runtime. For example:

```python
x = 3.14  # Python infers x is float
```

You don't explicitly declare `float x = 3.14` like in statically-typed languages (e.g., Java or C++).

> Tools like **type hints** (`x: int = 5`) help with optional static typing in Python.

#### 3. ⚖️ `is` vs `==`

#### ✅ `==` → **Value equality**
Checks **whether two values are equal**, regardless of whether they are the **same object** in memory.

```python
a = [1, 2]
b = [1, 2]
print(a == b)  # ✅ True (values are same)
```

#### ✅ `is` → **Identity check**
Checks **if two variables point to the same object** (i.e., same memory address).

```python
print(a is b)  # ❌ False (different objects)
```

#### 🔬 Exception:
For small integers and interned strings, Python may return `True` for both due to optimizations.

```python
x = 100
y = 100
print(x is y)  # True (due to interning)
```

> 🔍 **Interview Tip**: Use `is` for **`None` checks**, not for comparing values.
```python
if x is None:
    pass  # ✅ Good
```

#### 4. 🔄 **Type Casting**

#### 📌 Definition:
Converting a value from one type to another using **built-in functions**.

#### ✅ Examples:

```python
int("5")        # 5
float("3.14")   # 3.14
str(100)        # "100"
list("abc")     # ['a', 'b', 'c']
```

#### 🔥 Caution:
Unsafe casts will raise exceptions:

```python
int("hello")  # ❌ ValueError
```

> 🔍 **Use Case**: Type casting is essential when reading input from external sources like files or APIs, where data comes in as strings.


#### ✅ Summary Table

| Concept          | Description                              | Example                       |
|------------------|------------------------------------------|-------------------------------|
| Dynamic Typing   | Type decided at runtime                  | `x = "hello"`                 |
| Type Inference   | Python guesses the type                  | `y = 42` → int                |
| `==`             | Checks value equality                    | `[1, 2] == [1, 2] → True`     |
| `is`             | Checks identity (same object?)           | `[1, 2] is [1, 2] → False`    |
| Type Casting     | Converting between types                 | `int("42") → 42`              |

### Q3: What are closures in Python?

**Answer:**
A closure is a function object that remembers values in enclosing scopes even if they are not in memory anymore.

```python
def outer(x):
    def inner(y):
        return x + y
    return inner

closure_fn = outer(10)
print(closure_fn(5))  # Output: 15
```

**Scenario:**

### Q4: What is a decorator in Python? How do you use it practically?

A **decorator** is a function that takes another function (or class) and **extends or modifies** its behavior *without changing the source code*.

They rely on:
- **Functions being first-class objects**
- **Closures**


#### ✅ 1. **Function Decorators (without arguments)**

```python
def logger(func):
    def wrapper(*args, **kwargs):
        print(f"Calling {func.__name__}")
        return func(*args, **kwargs)
    return wrapper

@logger
def greet(name):
    print(f"Hello, {name}")

greet("Alice")
```

🧠 **Interview Insight**:  
Use decorators to add cross-cutting concerns like **logging**, **authentication**, **timing**, etc.

#### 🎯 2. **Function Decorators (with arguments)**

To make a decorator accept arguments, you need a **decorator factory** (a function that returns a decorator).

```python
def repeat(times):
    def decorator(func):
        def wrapper(*args, **kwargs):
            for _ in range(times):
                func(*args, **kwargs)
        return wrapper
    return decorator

@repeat(3)
def say_hello():
    print("Hello!")

say_hello()
```

🧠 **Interview Tip**: This shows your grasp of *higher-order functions* and nested closures.

#### 🔗 3. **Chaining Multiple Decorators**

You can apply multiple decorators by stacking them.

```python
def bold(func):
    def wrapper(*args, **kwargs):
        return f"<b>{func(*args, **kwargs)}</b>"
    return wrapper

def italic(func):
    def wrapper(*args, **kwargs):
        return f"<i>{func(*args, **kwargs)}</i>"
    return wrapper

@bold
@italic
def greet():
    return "Hello"

print(greet())  # Output: <b><i>Hello</i></b>
```

📌 **Order matters**: Inner decorator runs first (`@italic` runs before `@bold` in the above example).

#### 🧱 4. **Class Decorators (Basic)**

Decorators can also be implemented as classes using `__call__`.

```python
class CountCalls:
    def __init__(self, func):
        self.func = func
        self.count = 0

    def __call__(self, *args, **kwargs):
        self.count += 1
        print(f"Call #{self.count}")
        return self.func(*args, **kwargs)

@CountCalls
def say_hi():
    print("Hi!")

say_hi()
say_hi()
```

🧠 **When to use**: Class decorators are great when you want to **preserve state** between function calls (like caching or counting calls).

#### ⚙️ Pro Interview Insight

- Use `functools.wraps(func)` inside your wrapper to preserve metadata:
```python
from functools import wraps

def decorator(func):
    @wraps(func)
    def wrapper(*args, **kwargs):
        return func(*args, **kwargs)
    return wrapper
```

### Q5: What are iterators in Python?

**Answer:**
In Python, the iterator protocol is what powers for loops, comprehensions, and generators.

✅ Interview Explanation:
__iter__() returns the iterator object itself.

__next__() returns the next value from the iterator.

When no more data is available, __next__() should raise a StopIteration.

```python
class Count:
    def __init__(self, start, end):
        self.current = start
        self.end = end

    def __iter__(self):
        return self

    def __next__(self):
        if self.current >= self.end:
            raise StopIteration
        val = self.current
        self.current += 1
        return val

c = Count(1, 4)
for num in c:
    print(num)
# Output: 1, 2, 3
```
**Scenario:**
You need to read a huge file line-by-line without loading the whole thing into memory — using an iterator allows lazy evaluation.

🧠 **Interviewer might ask:** How does for loop internally work?
🔑 Answer: It calls iter(obj) → gets an iterator → repeatedly calls next() until StopIteration.

### Q6: What are generators? How are they different from iterators?

**Answer:**
Generators are a simpler way to create iterators using `yield`.

```python
def countdown(n):
    while n > 0:
        yield n
        n -= 1
```
`yield` turns a function into a generator. It remembers the function’s state between yields.

```python
def simple_gen():
    yield 1
    yield 2

for i in simple_gen():
    print(i)
```
**Scenario:**
You're processing a large CSV file. A generator can yield one parsed row at a time, avoiding high memory usage.

`yield from` delegates part of the generator to another iterable/generator.

```python
def outer():
    yield 1
    yield from [2, 3]
    yield 4

print(list(outer()))  # [1, 2, 3, 4]
```
**Scenario:**
Use `yield from` to flatten nested generators or simplify delegation.

### Q7: How does exception handling work in Python?

**Answer:**
Exception handling in Python is done using `try`, `except`, `else`, and `finally` blocks:

```python
try:
    result = 10 / 0
except ZeroDivisionError:
    print("Can't divide by zero")
else:
    print("No error occurred")
finally:
    print("This block always runs")
```

- `try`: Wraps code that may raise an exception
- `except`: Catches and handles specific exceptions
- `else`: Runs if no exception is raised
- `finally`: Always executes (for cleanup or logging)

**Scenario:**
You're calling a third-party API and want to handle connection timeouts gracefully, log the issue, and retry if needed.

### Q8: Difference between `for` and `while` loops in Python?

**Answer:**
- `for` loops are used when you know ahead of time how many iterations you want.
- `while` loops are used when the iteration depends on a condition rather than a range or list.

```python
for i in range(3):
    print(i)  # prints 0,1,2

x = 0
while x < 3:
    print(x)
    x += 1
```

**Scenario:**
Use a `for` loop to process items in a list. Use a `while` loop to poll an API every few seconds until you get a success response.

### Q11. What are Command-Line Arguments?

Command-line arguments are inputs **passed to a script when it's run from the terminal**.

> Example:  
```bash
python script.py arg1 arg2 arg3
```

These `arg1`, `arg2`, `arg3` are command-line arguments.

#### 🛠️ Accessing Command-Line Arguments in Python

#### ✅ Using `sys.argv` (Basic)

```python
import sys

print("Script name:", sys.argv[0])       # script.py
print("Arguments passed:", sys.argv[1:]) # ['arg1', 'arg2']
```

- `sys.argv` is a list of strings.
- `sys.argv[0]` is always the script name.
- All other items are arguments.

#### ✅ Using `argparse` (Recommended for real-world use)

```python
import argparse

parser = argparse.ArgumentParser(description="Add two numbers")
parser.add_argument("a", type=int, help="First number")
parser.add_argument("b", type=int, help="Second number")
args = parser.parse_args()

print("Sum =", args.a + args.b)
```

Run from terminal like:
```bash
python script.py 5 10
# Output: Sum = 15
```

#### ✨ Benefits of `argparse`:
- Adds help messages
- Validates types
- Handles optional arguments, flags, etc.

#### 🆚 `sys.argv` vs `argparse`

| Feature            | `sys.argv`         | `argparse`            |
|--------------------|--------------------|------------------------|
| Simplicity         | Simple              | More powerful & verbose |
| Type handling      | Manual              | Automatic               |
| Help/Usage message | ❌ No                | ✅ Yes                  |
| Error handling     | Manual              | Built-in                |


#### 🧠 Interview Tip:

> “Command-line arguments make scripts reusable and flexible. `argparse` is preferred in production-grade scripts due to its clarity, error handling, and documentation.”

### Q12. Example with optional flags and multiple inputs for cmd args using argparse

#### 🧪 Script Example: File Processor with Optional Flags

Let’s create a script that:

- Takes a list of input files
- Accepts an optional output filename (`--output`)
- Has a `--verbose` flag to print extra info


#### ✅ Code using `argparse`:

```python
import argparse

parser = argparse.ArgumentParser(description="Process text files.")

# Positional argument (multiple files)
parser.add_argument("files", nargs="+", help="List of input files")

# Optional argument with a value
parser.add_argument("--output", "-o", help="Output file name", default="result.txt")

# Optional boolean flag
parser.add_argument("--verbose", "-v", action="store_true", help="Enable verbose output")

args = parser.parse_args()

# Display inputs if verbose mode is on
if args.verbose:
    print(f"Processing {len(args.files)} files...")
    print(f"Output will be saved to: {args.output}")

# Simulate processing
with open(args.output, "w") as out_file:
    for file in args.files:
        if args.verbose:
            print(f"Reading {file}...")
        with open(file, "r") as f:
            out_file.write(f.read())
            out_file.write("\n")  # Separate content from each file
```

#### 🧪 How to Run:

```bash
python process_files.py file1.txt file2.txt --output merged.txt --verbose
```

- `file1.txt` and `file2.txt` → input files
- `--output merged.txt` → optional output filename
- `--verbose` → prints progress messages

#### 🧠 Interview Insight:

> "In production-level scripts, we use optional arguments to make tools more user-friendly and flexible. Flags like `--verbose` help in debugging, while defaults ensure scripts don't fail unnecessarily."

### Q14: Explain the following topics.: `def`,Lambdas,First class functions,default and keyword args,function introspection,scope behavior

#### ✅ 1. `def` — **Defining Functions**
This is the standard way to define functions in Python.

```python
def greet(name):
    return f"Hello, {name}!"
```

🧠 **Interview Insight**:  
Functions in Python are objects — even those defined with `def`. This makes Python a functional programming-friendly language.

#### 🕵️‍♀️ 2. **Anonymous Functions** (Lambdas)

Used for short, one-liner functions without a name.

```python
square = lambda x: x * x
print(square(5))  # 25
```

✅ Best for use as arguments to functions like `map`, `filter`, or `sorted`.

> 🔍 Interview Tip: Lambdas are limited — they can only contain expressions, not statements.

#### 🧠 3. **First-Class Functions**

In Python, functions are first-class citizens:
- Assigned to variables
- Passed as arguments
- Returned from other functions

```python
def shout(text):
    return text.upper()

def greet(func):
    return func("hello")

print(greet(shout))  # HELLO
```

#### 🧵 4. **Default Arguments**

Used to assign default values to parameters if no value is passed.

```python
def greet(name="Guest"):
    return f"Hello, {name}"
```

🛑 **Mutable default trap**:
```python
def add_item(item, my_list=[]):  # BAD
    my_list.append(item)
    return my_list
```

Always use `None` as a default and initialize inside:
```python
def add_item(item, my_list=None):
    if my_list is None:
        my_list = []
```

#### 🔑 5. **Keyword-Only Arguments** (`*` marker)

Forces callers to use **keywords** instead of positional arguments.

```python
def connect(host, *, timeout=5):
    print(f"Connecting to {host} with timeout {timeout}")
```

Now `timeout` **must** be passed as a keyword:
```python
connect("localhost", timeout=10)  # ✅
connect("localhost", 10)          # ❌ TypeError
```

##### 🔍 6. **Function Introspection**

You can inspect a function's metadata:

```python
def example(a, b=10): pass

print(example.__name__)     # 'example'
print(example.__defaults__) # (10,)
print(example.__code__.co_varnames)  # ('a', 'b')
```

For deeper inspection:
```python
import inspect
print(inspect.signature(example))
```

#### 🌍 7. **Scope Behavior**

Python follows **LEGB Rule** for name resolution:
- **L**ocal
- **E**nclosing
- **G**lobal
- **B**uilt-in

```python
x = "global"

def outer():
    x = "enclosing"
    def inner():
        x = "local"
        print(x)
    inner()

outer()  # local
```

Use `global` and `nonlocal` carefully:
```python
def outer():
    count = 0
    def inner():
        nonlocal count
        count += 1
```

### Q1: What is the difference between a class, metaclass, and abstract class in Python?

**Answer:**
- **Class**: Blueprint for creating objects (instances).
- **Metaclass**: Class of a class; controls class creation behavior <refer Q16 for detailed ans>
- **Abstract Class**: A class that can't be instantiated directly and must be subclassed.

```python
from abc import ABC, abstractmethod

class Animal(ABC):
    @abstractmethod
    def sound(self):
        pass

class Dog(Animal):
    def sound(self):
        return "Bark"
```

**Scenario:**

### Q11.Explain the difference between `self` and `cls` in Python? Also, when would you use a static method, a class method, and an instance method?

#### 🔹 **`self` vs `cls`**

| Keyword | Refers to                    | Used in                | Access to | Purpose |
|---------|------------------------------|------------------------|------------|---------|
| `self`  | The **instance** of the class | Instance methods        | Instance variables & methods | To operate on data unique to each object |
| `cls`   | The **class** itself          | Class methods (`@classmethod`) | Class variables & other class methods | To operate on the class or factory patterns |


#### ✅ **Method Types in Python**

| Type             | Decorator        | First Arg | Can Access          | When to Use |
|------------------|------------------|-----------|----------------------|-------------|
| **Instance Method** | _(none)_       | `self`    | Instance & class data | When behavior depends on instance state |
| **Class Method**    | `@classmethod` | `cls`     | Class data only       | When you need to create or modify the class itself |
| **Static Method**   | `@staticmethod`| —         | None by default       | When logic is related to the class but doesn’t access class/instance data |


#### 🧪 **Examples**

```python
class Circle:
    pi = 3.14159  # class variable
    
    def __init__(self, radius):
        self.radius = radius

    # Instance method
    def area(self):
        return self.pi * self.radius ** 2

    # Class method
    @classmethod
    def from_diameter(cls, diameter):
        radius = diameter / 2
        return cls(radius)

    # Static method
    @staticmethod
    def unit():
        return Circle(1)
```

```python
c1 = Circle(3)
print(c1.area())                      # Uses self => 28.27
c2 = Circle.from_diameter(10)        # Uses cls => Circle(radius=5)
c3 = Circle.unit()                   # Uses staticmethod => Circle(radius=1)
```

#### ⚠️ **Common Interview Gotchas**

> **Q:** Can a `@staticmethod` access instance or class variables?

**A:** No, unless you pass them explicitly. It behaves like a plain function **inside a class namespace**.

> **Q:** Why use `@classmethod` over `@staticmethod`?

**A:** Use `@classmethod` when you need to **access or modify class-level data** or implement **factory methods** that return instances of the class.


#### 🧠 Tip:
Think of:
- `self` → “Me” (the current object)
- `cls` → “We” (the whole class)
- `staticmethod` → “Detached helper”

### Q12. What is inheritance in Python? Can you explain with examples and mention its types?

#### ✅ **Definition:**

**Inheritance** allows a class (called the **child/subclass**) to inherit **attributes and methods** from another class (called the **parent/superclass**).  
It promotes **code reusability**, **extensibility**, and **logical hierarchy** in object-oriented design.


#### 🔧 **Basic Syntax:**

```python
class Parent:
    def greet(self):
        print("Hello from Parent!")

class Child(Parent):  # Inheriting Parent
    def say_hi(self):
        print("Hi from Child!")

c = Child()
c.greet()  # Inherited from Parent
c.say_hi()  # Defined in Child
```

#### 🔍 **Types of Inheritance in Python:**

| Type             | Description                                   | Example                       |
|------------------|-----------------------------------------------|-------------------------------|
| **Single**       | One child, one parent                         | `class B(A):`                 |
| **Multiple**     | One child, multiple parents                   | `class C(A, B):`              |
| **Multilevel**   | Inheritance through a chain                   | `class C(B):` from `class B(A):` |
| **Hierarchical** | Multiple children from one parent             | `class B(A), class C(A):`     |
| **Hybrid**       | A combination of any of the above             | Mix of multilevel + multiple  |


#### 🧠 **Example: Multilevel + Method Overriding**

```python
class Animal:
    def speak(self):
        print("Animal speaks")

class Dog(Animal):
    def speak(self):
        print("Dog barks")  # Overriding

class Puppy(Dog):
    pass

p = Puppy()
p.speak()  # Dog barks (inherited + overridden)
```

#### 📌 **Method Overriding**

You can **redefine** a method in the child class using the same name as in the parent class.

#### 🔄 **Using `super()`**

Use `super()` to call the parent class method from the child.

```python
class Dog(Animal):
    def speak(self):
        super().speak()
        print("Dog barks")
```

#### 🚧 **Important Interview Points:**

> **Q:** Does Python support multiple inheritance?  
**A:** Yes, Python supports multiple inheritance.

> **Q:** How does Python resolve method names in multiple inheritance?  
**A:** Using **MRO** (Method Resolution Order) — it follows the **C3 Linearization** algorithm.

#### 🔍 **Example of Multiple Inheritance:**

```python
class A:
    def show(self): print("A")

class B:
    def show(self): print("B")

class C(A, B):  # Multiple inheritance
    pass

c = C()
c.show()  # Output: A (as per MRO)
```

### Q13. What are the SOLID Principles?

**S.O.L.I.D** is an acronym for:
- **S**: Single Responsibility Principle (SRP)
- **O**: Open/Closed Principle (OCP)
- **L**: Liskov Substitution Principle (LSP)
- **I**: Interface Segregation Principle (ISP)
- **D**: Dependency Inversion Principle (DIP)

#### 1. 🧹 **Single Responsibility Principle (SRP)**  
**Each class should have only one reason to change.**

✅ **Do one thing, and do it well.**

**Bad Example:**
```python
class User:
    def __init__(self, name):
        self.name = name

    def save_to_db(self):
        pass  # DB logic

    def send_email(self):
        pass  # Email logic
```

📛 This class handles **user data**, **DB logic**, and **emailing** — too much!

**Good Example (SRP Applied):**
```python
class User:
    def __init__(self, name):
        self.name = name

class UserDB:
    def save(self, user):
        pass  # DB logic

class EmailService:
    def send_email(self, user):
        pass  # Email logic
```

🔧 **Interview Tip**: "SRP helps make code modular and testable. Each responsibility goes into its own class or module."

#### 2. 🚪 **Open/Closed Principle (OCP)**  
**Software entities should be open for extension, but closed for modification.**

✅ You should be able to **add new behavior** without changing existing code.

**Bad Example:**
```python
def get_discount(type):
    if type == 'silver':
        return 0.1
    elif type == 'gold':
        return 0.2
```

Every time we add a new type, we modify this function — violates OCP.

**Good Example (OCP Applied):**
```python
class Discount:
    def get_discount(self):
        return 0

class SilverDiscount(Discount):
    def get_discount(self):
        return 0.1

class GoldDiscount(Discount):
    def get_discount(self):
        return 0.2
```

Now we can **extend** by creating new classes without touching the existing ones.

#### 3. 🔄 **Liskov Substitution Principle (LSP)**  
**Subclasses should be replaceable with their base class without altering correctness.**

✅ A derived class **must not break** the parent class behavior.

**Bad Example:**
```python
class Bird:
    def fly(self):
        pass

class Ostrich(Bird):
    def fly(self):
        raise Exception("Ostriches can't fly")
```

📛 Violates LSP — `Ostrich` is not truly substitutable for `Bird`.

**Good Example:**
```python
class Bird:
    pass

class FlyingBird(Bird):
    def fly(self):
        pass

class Sparrow(FlyingBird):
    def fly(self):
        print("Flying")

class Ostrich(Bird):
    pass  # No fly method
```

📌 Subclasses now behave consistently.


#### 4. 🔌 **Interface Segregation Principle (ISP)**  
**Clients shouldn’t be forced to depend on interfaces they don’t use.**

✅ Break big interfaces into smaller ones.

---

**Bad Example:**
```python
class Machine:
    def print(self): pass
    def scan(self): pass
    def fax(self): pass
```

What if a machine only supports printing?

---

**Good Example:**
```python
class Printer:
    def print(self): pass

class Scanner:
    def scan(self): pass

class Fax:
    def fax(self): pass

class SimplePrinter(Printer):
    def print(self): print("Printing...")
```

🔧 This lets us mix and match functionality based on the needs.

#### 5. 🔁 **Dependency Inversion Principle (DIP)**  
**High-level modules shouldn’t depend on low-level modules. Both should depend on abstractions.**

✅ Depend on **interfaces**, not concrete implementations.

**Bad Example:**
```python
class MySQLDatabase:
    def connect(self):
        pass

class UserService:
    def __init__(self):
        self.db = MySQLDatabase()
```

Tightly coupled — what if we switch to PostgreSQL?

**Good Example (DIP Applied):**
```python
class Database:
    def connect(self):
        pass

class MySQLDatabase(Database):
    def connect(self):
        print("MySQL connected")

class UserService:
    def __init__(self, db: Database):
        self.db = db
```

Now `UserService` can work with **any kind of DB** (as long as it implements `Database`).

#### 🔚 Summary Table

| Principle | Focus                         | Benefit                      |
|----------|------------------------------|-----------------------------|
| SRP       | One responsibility per class | Simplicity, testability     |
| OCP       | Open to extend, closed to modify | Extensibility             |
| LSP       | Subclasses respect parent    | Substitutability, correctness |
| ISP       | Small, specific interfaces   | Flexibility, clarity        |
| DIP       | Abstractions > concrete deps | Decoupling, scalability     |

### Q14. What is `__slots__` in Python?

By default, Python uses a flexible `__dict__` to store instance attributes. This allows dynamic attribute assignment, but it consumes more memory.

`__slots__` is a way to **tell Python exactly what attributes** an object will have, so it can use a more memory-efficient internal representation.

##### 🧠 Why use `__slots__`?

| Benefit           | Explanation |
|------------------|-------------|
| ✅ **Memory optimization** | Avoids creating `__dict__` for each instance |
| ✅ **Faster attribute access** | Because Python doesn't need to look things up in a dictionary |
| 🚫 **Limits flexibility** | You can't add new attributes that aren't listed in `__slots__` |

#### 📦 Example:

```python
class Person:
    __slots__ = ['name', 'age']  # Allowed attributes

    def __init__(self, name, age):
        self.name = name
        self.age = age

p = Person("Pratyusha", 27)
p.name  # Works

p.city = "Hyderabad"  # ❌ AttributeError: 'Person' object has no attribute 'city'
```

Without `__slots__`, the above would work — but also use more memory.

#### 🔬 Memory Comparison:

```python
import sys

class NoSlots:
    def __init__(self):
        self.x = 1
        self.y = 2

class WithSlots:
    __slots__ = ['x', 'y']
    def __init__(self):
        self.x = 1
        self.y = 2

print(sys.getsizeof(NoSlots()))   # ~56 bytes
print(sys.getsizeof(WithSlots())) # ~40 bytes (smaller!)
```

#### ⚠️ When **Not** to Use `__slots__`:

- If you need **dynamic attributes** (adding attrs at runtime)
- If you're using **multiple inheritance** (slots can complicate MRO)
- If you're building **flexible data models**

#### 🧠 Interview Tip:

> "`__slots__` is great for performance tuning, especially when you're creating many objects with predictable attributes. However, it trades flexibility for efficiency, so it’s best used in controlled environments like game engines, microservices, or data-heavy applications."

### Q16. What Are Metaclasses in Python?
#### 💡 **Definition:**
A **metaclass** in Python is a **"class of a class."**  
Just like a class defines how an **object** behaves, a **metaclass defines how a class behaves**.

In simpler terms:

> ✅ *Objects are instances of classes*  
> ✅ *Classes themselves are instances of metaclasses*

#### 🤔 **Why Use Metaclasses?**

Metaclasses allow you to:
- Control **class creation**
- Enforce **coding standards**
- Automatically **modify class attributes or methods**
- Add **custom logic** at the time the class is defined (not when objects are created)

#### 🧪 **Basic Example Without Metaclasses**
```python
class Person:
    def __init__(self, name):
        self.name = name

p = Person("Alice")
print(type(p))         # <class '__main__.Person'>
print(type(Person))    # <class 'type'>
```

Even `Person` is an instance of `type`. So `type` is the **default metaclass** in Python.

### 🔧 **Creating a Custom Metaclass**

```python
# Step 1: Create a metaclass
class MyMeta(type):
    def __new__(cls, name, bases, dct):
        print(f"Creating class: {name}")
        dct['created_by'] = 'Metaclass Magic ✨'
        return super().__new__(cls, name, bases, dct)

# Step 2: Use it in a class
class MyClass(metaclass=MyMeta):
    pass

print(MyClass.created_by)  # Output: Metaclass Magic ✨
```

🌀 Here’s what’s happening:
- `MyMeta.__new__()` is triggered **when the class `MyClass` is defined**, not when an instance is created.
- We dynamically added a new attribute `created_by`.

#### 🔎 **When Would You Use This in Real Life?**

#### ✅ **Enforcing Class Structure:**
```python
class InterfaceChecker(type):
    def __new__(cls, name, bases, dct):
        if 'do_work' not in dct:
            raise TypeError("All subclasses must implement do_work()")
        return super().__new__(cls, name, bases, dct)

class Base(metaclass=InterfaceChecker):
    pass

# This will fail
class Worker(Base):
    pass  # ❌ Raises error: do_work not implemented
```

This pattern is common when creating **plugin architectures, frameworks, or enforcing interfaces**.

#### ⚠️ **Risks and Complexity:**
- Can make your code **harder to read and debug**
- Often **overkill** unless you're building a **framework or DSL**
- Use metaclasses **only when class behavior truly needs control**


#### 🎤 **Interview Tip:**

> "Metaclasses allow developers to intercept and customize class creation in Python. While powerful, they're best used in scenarios where enforcing structure, injecting behavior, or auto-generating boilerplate at class-definition time adds significant value — like in ORMs, serializers, or custom framework components."

### Q17. What is a Descriptor?

A **descriptor** is any object that defines at least one of the following methods:

- `__get__(self, instance, owner)` – for reading the attribute
- `__set__(self, instance, value)` – for writing to the attribute
- `__delete__(self, instance)` – for deleting the attribute

Any object that implements any of these methods is called a **descriptor**.

##### 📦 Descriptor in Action – Example:

```python
class UpperCaseDescriptor:
    def __get__(self, instance, owner):
        return instance._name.upper()

    def __set__(self, instance, value):
        instance._name = value

class Person:
    name = UpperCaseDescriptor()  # descriptor

    def __init__(self, name):
        self._name = name

p = Person("pratyusha")
print(p.name)  # Output: PRATYUSHA
```

Here, the descriptor intercepts attribute access and transforms the value.

#### 🧠 Types of Descriptors:

| Type                 | Methods Implemented   | Behavior                        |
|----------------------|-----------------------|---------------------------------|
| **Data Descriptor**  | `__get__`, `__set__` or `__delete__` | Takes precedence over instance `__dict__` |
| **Non-data Descriptor** | Only `__get__`     | Used only if not found in instance `__dict__` |

✅ **Data Descriptors override instance attributes**, while non-data descriptors don’t.

#### 💡 Real-World Use Cases:

- `@property` uses descriptors under the hood
- `staticmethod` and `classmethod`
- ORM field mapping (like Django models)
- Input validation

#### ⚙️ Descriptor vs Property?

**@property** is a **high-level descriptor** that allows you to define getter/setter logic without writing a full class.

```python
class Person:
    def __init__(self, name):
        self._name = name

    @property
    def name(self):
        return self._name.upper()

    @name.setter
    def name(self, value):
        self._name = value
```

Under the hood, this is syntactic sugar for a descriptor class!

## 🔥 Interview Insight:

> “Descriptors provide fine-grained control over attribute access. They're useful when you want to enforce rules or transformations consistently across instances — like custom validators, caching, logging, etc.”

### Q18. `str()` vs `repr()`: What’s the difference?

Both `str()` and `repr()` are used to **get string representations** of objects, but they serve different purposes:

| Aspect        | `str()`                                    | `repr()`                                      |
|---------------|---------------------------------------------|------------------------------------------------|
| **Goal**      | Human-readable (informal)                  | Developer/debug-friendly (formal/unambiguous)  |
| **Audience**  | End users                                  | Developers                                     |
| **Used by**   | `print()`, `str(obj)`                      | `repr(obj)`, interactive shell, debugging      |
| **Format**    | Readable and clean                         | Often includes type info or quotes             |
| **Fallback**  | If `__str__` not defined → uses `__repr__` | If `__repr__` not defined → defaults to `<...>` |

##### 📦 Example:

```python
class Book:
    def __init__(self, title):
        self.title = title

    def __str__(self):
        return f"Book Title: {self.title}"

    def __repr__(self):
        return f"Book('{self.title}')"

b = Book("Harry Potter")
print(str(b))   # Book Title: Harry Potter → for users
print(repr(b))  # Book('Harry Potter')    → for devs
```

#### 🔁 Internally:
- `str(obj)` calls `obj.__str__()`
- `repr(obj)` calls `obj.__repr__()`
- If `__str__()` is **not defined**, `str(obj)` falls back to `__repr__()`


##### 🔍 Interview Tip:

> "`str` is meant to be readable. `repr` is meant to be unambiguous — ideally, the output of `repr(obj)` should be valid Python code to recreate the object when possible."

Example:

```python
from datetime import datetime
now = datetime.now()

print(str(now))   # 2025-04-17 14:22:41.621422
print(repr(now))  # datetime.datetime(2025, 4, 17, 14, 22, 41, 621422)
```

### Q19. What are the OOPS Principles?

#### 🧱 1. **Encapsulation**  
**Hiding internal state** and requiring all interaction to be performed through an object's methods.

##### 🔹 In Python:
We use **private/protected** attributes (`_var`, `__var`) and **getters/setters**.

```python
class Account:
    def __init__(self, balance):
        self.__balance = balance  # private

    def deposit(self, amount):
        if amount > 0:
            self.__balance += amount

    def get_balance(self):
        return self.__balance
```

✅ **Benefit**: Prevents direct modification, adds validation, and protects object integrity.


#### 🧱 2. **Abstraction**  
**Hiding complex implementation** details and showing only the **essential features**.

##### 🔹 In Python:
Achieved using **abstract base classes (ABC)** or by simply exposing minimal public API.

```python
from abc import ABC, abstractmethod

class Animal(ABC):
    @abstractmethod
    def sound(self):
        pass

class Dog(Animal):
    def sound(self):
        return "Woof!"
```

✅ **Benefit**: Focuses on *what* an object does, not *how*. Enables easier changes & usage.


## 🧱 3. **Inheritance**  
One class **inherits** attributes and methods from another.

```python
class Vehicle:
    def start(self):
        print("Vehicle starting...")

class Car(Vehicle):
    def start(self):
        print("Car engine starting...")

c = Car()
c.start()  # Car engine starting...
```

✅ **Benefit**: Code reuse, clear hierarchy, easier maintenance.

## 🧱 4. **Polymorphism**  
**Same interface**, different implementation. Allows functions/methods to work with different types.

```python
class Cat:
    def speak(self):
        return "Meow"

class Dog:
    def speak(self):
        return "Woof"

def make_sound(animal):
    print(animal.speak())

make_sound(Cat())  # Meow
make_sound(Dog())  # Woof
```

✅ **Benefit**: Flexibility. Easily extend behavior without modifying existing code.

##### 🧠 Interview Tip:
> “OOP principles help structure code that's easier to understand, extend, and maintain. Python supports all these principles, though it allows flexibility through duck typing and dynamic typing.”

### Q21. Can you explain Python's `dataclass` and how it simplifies the creation of classes?

**Answer:**  
Yes, a `dataclass` in Python is a decorator that was introduced in Python 3.7. It simplifies the creation of classes that are primarily used to store data without having to write boilerplate code like `__init__`, `__repr__`, `__eq__`, etc.

#### Key Features of `dataclass`:
1. **Automatic Generation of Special Methods:**
   - When you use the `@dataclass` decorator, Python automatically generates common methods like:
     - `__init__`: The constructor to initialize the data attributes.
     - `__repr__`: A readable string representation of the object, which is especially useful for debugging.
     - `__eq__`: A comparison method to compare two instances of the dataclass.
     - `__hash__`: If applicable, generates a hash method for immutable dataclasses.
     - `__post_init__`: A special method that is called after the `__init__` method, used for any additional initialization logic.

2. **Simplification of Data Classes:**
   - Dataclasses are typically used for objects that store state, and the decorator takes care of much of the boilerplate code, making the class definition cleaner and more concise.

#### Example:

```python
from dataclasses import dataclass

@dataclass
class Person:
    name: str
    age: int
    city: str = "Unknown"  # default value

# Creating an object of the class
person = Person(name="Alice", age=30)

# Accessing attributes
print(person.name)  # Output: Alice
print(person.age)   # Output: 30
print(person.city)  # Output: Unknown (default value)

# Using the auto-generated __repr__ method
print(person)  # Output: Person(name='Alice', age=30, city='Unknown')
```

#### When to Use `dataclass`:
- It's best used when you have classes whose primary purpose is to store data attributes and you don't need complex methods or behavior.
- The `dataclass` is ideal for representing immutable objects (like `namedtuple`), but it can also handle mutable objects when you use the `frozen=True` parameter.

#### `frozen=True` (Immutability):
If you want to make a dataclass immutable (meaning that its attributes cannot be changed after instantiation), you can use the `frozen=True` parameter:

```python
@dataclass(frozen=True)
class Point:
    x: int
    y: int

point = Point(1, 2)
# point.x = 3  # This will raise a FrozenInstanceError
```

#### Customizing the `dataclass`:
You can add additional behavior to the class through `__post_init__` and other methods, but the goal of `dataclass` is to reduce the repetitive work of writing standard methods for a data container.

**Follow-up Question:**  
How would you handle validation of attributes in a dataclass?

**Answer:**  
You can handle validation in a `dataclass` by using the `__post_init__` method, which is automatically called after the `__init__` method. In this method, you can include any logic to validate the values of the attributes.

#### Example of Validation:

```python
from dataclasses import dataclass

@dataclass
class Employee:
    name: str
    age: int

    def __post_init__(self):
        if self.age < 18:
            raise ValueError("Age must be at least 18")
        if not self.name:
            raise ValueError("Name cannot be empty")

# Valid case
emp = Employee(name="John", age=30)

# Invalid case: Raises ValueError
emp = Employee(name="Alice", age=16)  # ValueError: Age must be at least 18
```

In this example, the `__post_init__` method checks the `age` and `name` attributes for validity and raises exceptions if the conditions are not met.

### Q22. Difference between methods and functions. Why is `len()` and `append()` a function and method respectively? Since both are acting on a list, why aren't they the same?

#### 🔹 **Core Difference: Function vs Method**

| Aspect         | **Function**                | **Method**                        |
|----------------|-----------------------------|-----------------------------------|
| **Definition** | A function is a block of code that performs a task and is defined using `def` or a built-in function like `len()`. | A method is a function that is associated with an object (i.e., bound to a class/instance). |
| **How it's called** | Called directly: `len(some_list)` | Called on an object: `some_list.append(item)` |
| **Belongs to** | Not tied to any object/class (though it may work *with* objects). | Belongs to a class and is invoked using dot (`.`) notation. |
| **First argument** | Does **not** implicitly receive `self`. | Always receives `self` (the object) implicitly as the first argument. |


#### 🔹 **Examples:**

#### ✅ `len()` → **Function**
- **Built-in** function in Python.
- Not called on an object.
- Works with various types (list, str, tuple, etc.).

```python
my_list = [1, 2, 3]
print(len(my_list))  # Output: 3
```

> Here, `len()` is not a method on the list. It internally calls `my_list.__len__()`, but it's still a standalone function.

#### ✅ `append()` → **Method**
- A method **defined on list objects**.
- Called with dot notation on a specific list instance.

```python
my_list = [1, 2, 3]
my_list.append(4)  # Modifies the list in-place
print(my_list)     # Output: [1, 2, 3, 4]
```

> Here, `append()` is a method that modifies the list it's called on — it's **bound** to `my_list`.

#### 🎯 Quick Tip to Identify:
- If you see `object.something()` → it's probably a **method**.
- If you see `something(object)` → it's a **function** (likely standalone or built-in).

#### 🧠 Interview Trick:
You can say:
> "`len()` is a **function** because it's defined at the module level and takes the object as an argument. `append()` is a **method** because it's defined inside the `list` class and is invoked on an instance of that class."

#### 🔁 **Nested Dict → Flat Dict**

#### ✅ Example:
```python
def flatten_dict(d, parent_key='', sep='.'):
    flat = {}
    for k, v in d.items():
        key = f"{parent_key}{sep}{k}" if parent_key else k
        if isinstance(v, dict):
            flat.update(flatten_dict(v, key, sep))
        else:
            flat[key] = v
    return flat

nested = {'a': {'b': {'c': 1}}, 'd': 2}
flat = flatten_dict(nested)
print(flat)  # {'a.b.c': 1, 'd': 2}
```

## 🔹 Section 5: Concurrency and Parallelism

### Q1: What is the difference between threading and multiprocessing in Python?

**Answer:**
- **Threading**: Runs multiple threads in a single process. Threads share memory space but are limited by the GIL (Global Interpreter Lock).
- **Multiprocessing**: Spawns separate processes with their own Python interpreter and memory space, bypassing GIL.

```python
from threading import Thread
import multiprocessing

# Threading example
def print_nums():
    for i in range(3):
        print(f"Thread {i}")

Thread(target=print_nums).start()

# Multiprocessing example
def print_proc():
    print("From process")

if __name__ == '__main__':
    multiprocessing.Process(target=print_proc).start()
```

**Scenario:** Use multiprocessing for CPU-bound tasks; threading for I/O-bound tasks.

### Q2: What are race conditions and how can you prevent them?

**Answer:** A race condition occurs when threads/processes access shared data simultaneously and the outcome depends on the order of execution.

**Prevention Tools:** Locks, Semaphores, Queues

```python
import threading

lock = threading.Lock()
shared_counter = 0

def increment():
    global shared_counter
    with lock:
        for _ in range(1000):
            shared_counter += 1

threads = [threading.Thread(target=increment) for _ in range(5)]
[thread.start() for thread in threads]
[thread.join() for thread in threads]
print(shared_counter)  # Expected: 5000
```

**Answer:** A deadlock occurs when two or more threads wait indefinitely for resources held by each other.

**Prevention:** Lock ordering, timeout acquisition, avoiding nested locks

### Q11. Compare how **stateless vs stateful** applications handle **session management in Flask**

##### 🧭 Scenario: A user logs into a website

We’ll implement the login process in two styles:

- **Stateless API** → using **JWT tokens**  
- **Stateful App** → using **Flask sessions**

#### ⚙️ 1. Stateless with JWT (Token-based Authentication)

Every request **contains all the information** the server needs. No data is stored on the server.

##### ✅ Flask Code (Stateless)
```python
from flask import Flask, request, jsonify
import jwt, datetime

app = Flask(__name__)
SECRET_KEY = "supersecret"

@app.route('/login', methods=['POST'])
def login():
    data = request.json
    if data['username'] == 'pratyusha' and data['password'] == '123':
        token = jwt.encode({
            'user': data['username'],
            'exp': datetime.datetime.utcnow() + datetime.timedelta(minutes=30)
        }, SECRET_KEY, algorithm='HS256')
        return jsonify({'token': token})
    return jsonify({'message': 'Invalid credentials'}), 401

@app.route('/dashboard', methods=['GET'])
def dashboard():
    token = request.headers.get('Authorization')
    if not token:
        return jsonify({'error': 'Token missing'}), 401

    try:
        data = jwt.decode(token.split()[1], SECRET_KEY, algorithms=['HS256'])
        return jsonify({'message': f"Welcome {data['user']}!"})
    except jwt.ExpiredSignatureError:
        return jsonify({'error': 'Token expired'}), 401
```

##### 🔁 Every request must send the token:
```bash
GET /dashboard
Authorization: Bearer <JWT_TOKEN>
```

**🧼 Server doesn't store any session data** = **Stateless**

#### 🗂️ 2. Stateful with Flask Session

The server **remembers the user** between requests using a cookie/session ID stored in memory or a database.

#### ✅ Flask Code (Stateful)
```python
from flask import Flask, session, request, redirect, jsonify

app = Flask(__name__)
app.secret_key = "secret"

@app.route('/login', methods=['POST'])
def login():
    data = request.json
    if data['username'] == 'pratyusha' and data['password'] == '123':
        session['user'] = data['username']
        return jsonify({'message': 'Logged in!'})
    return jsonify({'error': 'Invalid credentials'}), 401

@app.route('/dashboard', methods=['GET'])
def dashboard():
    if 'user' in session:
        return jsonify({'message': f"Welcome {session['user']}!"})
    return jsonify({'error': 'Please log in'}), 401
```

#### ✅ Behind the scenes:
- A session cookie (with a unique ID) is created after login
- Server keeps track of this session in memory or a file/database


#### 🧠 Summary: Key Differences

| Feature                  | Stateless (JWT)                      | Stateful (Flask session)              |
|--------------------------|--------------------------------------|---------------------------------------|
| **Server stores user?**   | ❌ No                                 | ✅ Yes                                 |
| **Scaling**              | ✅ Easy (no session memory)          | ❌ Needs sticky sessions / Redis      |
| **Authentication data**  | Sent on every request (JWT token)   | Stored in server memory or Redis      |
| **Used in**              | APIs, microservices                 | Web apps, workflows, multi-step forms |


#### 🧪 Which to use when?

- ✅ Use **stateful** when you need to **track user flows** like carts, bookings, forms

### Q1: Pickling and UnPickling:
These three terms — **pickling**, **serialization**, and **marshalling** — are closely related, often used interchangeably, but they do have nuanced differences depending on the context and language. Here's a breakdown:

---

### 🔹 **Serialization**

**Definition**:  
Serialization is the general process of converting a data structure or object into a format that can be stored (e.g., in a file or database) or transmitted (e.g., over a network) and later reconstructed.

**Purpose**:
- Saving the state of an object.
- Sending objects over a network.
- Storing objects in files or databases.

**Common formats**:
- Binary
- JSON
- XML
- Protocol Buffers, Avro, etc.

**In Python**:  
Serialization is commonly done using:
```python
import json

data = {"name": "Pratyusha", "age": 28}
json_str = json.dumps(data)  # serialize to JSON string
```

---

### 🔹 **Pickling (Python-specific)**

**Definition**:  
Pickling is Python’s way of serializing and de-serializing Python object structures using the `pickle` module.

**Pickle format**: A binary format that's specific to Python (not human-readable, and not language-agnostic).

**Example**:
```python
import pickle

data = {"name": "Pratyusha", "age": 28}
# Serialize
with open("data.pkl", "wb") as f:
    pickle.dump(data, f)

# Deserialize
with open("data.pkl", "rb") as f:
    loaded_data = pickle.load(f)
```

**⚠️ Note**: Pickled data is **not secure**. Only unpickle data from trusted sources — it can execute arbitrary code!

---

### 🔹 **Marshalling**

**Definition**:  
Marshalling is the process of preparing and converting an object or data so that it can be transferred across different environments (e.g., between different processes, machines, or languages).

**In many languages (like Java, Go, C)**:
- Marshalling refers to serialization with a focus on **inter-process communication (IPC)** or **remote procedure calls (RPC)**.

**In Python**:  
There's a module called `marshal`, but it’s **not recommended for general serialization**, as it's mainly used for Python's internal purposes (like `.pyc` files) and isn't guaranteed to be stable across versions.

Example:
```python
import marshal

data = {"name": "Pratyusha", "age": 28}
serialized = marshal.dumps(data)
deserialized = marshal.loads(serialized)
```

### Q1.What is Monkey Patching in Python? Can you explain it with an example and mention where it might be useful or risky?

#### ✅ **Definition:**

**Monkey patching** is a technique where you **dynamically modify or extend** classes or modules at **runtime** — usually by changing methods or attributes *without altering the source code*.

In Python, everything is dynamic, including functions — so it's totally legal (and sometimes useful) to override functionality *on the fly*.

#### 🔧 **Example – Monkey Patching a Class Method:**

```python
class Dog:
    def bark(self):
        print("Woof!")

# Original behavior
d = Dog()
d.bark()  # Woof!

# Monkey patching
def meow_instead(self):
    print("Meow?")

Dog.bark = meow_instead  # Replacing bark with meow

d.bark()  # Meow?
```
#### 🔍 **Use Cases:**
- **Testing/Mocking:** Override external APIs or services.
- **Hotfixes:** Fix a bug in a third-party library without modifying source code.
- **Quick Prototyping:** Swap behaviors dynamically for experiments.

#### ⚠️ **Risks:**
- **Unpredictable behavior** – especially in larger or multi-threaded systems.
- Can **break encapsulation** and lead to **maintenance headaches**.
- Makes code **hard to debug and understand** for others.

#### 🧪 **Use Case in Testing (Mocking Example):**

```python
import time

def slow_function():
    time.sleep(5)
    return "Done"

# Monkey patch time.sleep in a test
time.sleep = lambda x: None  # Bypasses delay

print(slow_function())  # Instant output: "Done"
```

#### 🧠 **Key Interview Takeaways:**

| Aspect           | Description                              |
|------------------|------------------------------------------|
| What             | Dynamic replacement of functions/methods |
| Where Used       | Testing, Hotfixes, Quick Prototyping     |
| Safe to use?     | Cautiously, especially in production     |
| Better Alternatives | Dependency injection, mocking libs like `unittest.mock` |

Let’s go through a **real-world example** of using **monkey patching to mock an external API call** during unit testing.

#### 🧪 **Scenario:**
Suppose you have a function that makes an API call to fetch weather data from a third-party service.

#### 🔧 **Production Code (`weather.py`)**
```python
import requests

def get_weather(city):
    response = requests.get(f"https://api.weather.com/{city}")
    return response.json()
```

Now during testing, you **don’t want to actually make that HTTP call** every time — it's slow, unreliable, and can consume API quota.

#### ✅ **Test with Monkey Patching**

```python
# test_weather.py
import weather

def fake_get(url):
    class FakeResponse:
        def json(self):
            return {"temp": "22°C", "condition": "Sunny"}
    return FakeResponse()

# Monkey patch requests.get
import requests
requests.get = fake_get  # ← Monkey patching here!

# Test
print(weather.get_weather("Delhi"))
# Output: {'temp': '22°C', 'condition': 'Sunny'}
```

#### ✅ Output:
You get a fake but controlled response **without hitting the real API** — super useful in unit testing.

#### 🧠 **Key Points to Bring Up in an Interview:**

- You replaced `requests.get` *at runtime* with a custom function (`fake_get`).
- This isolates your test from external dependencies (pure unit testing).
- It’s a quick way to test logic **without using libraries** like `unittest.mock` (though in production, those are safer).


#### 🔄 Safer Alternative (Recommended in Production):
Use `unittest.mock.patch`:

```python
from unittest.mock import patch

@patch('weather.requests.get')
def test_weather(mock_get):
    mock_get.return_value.json.return_value = {"temp": "22°C", "condition": "Sunny"}
    assert weather.get_weather("Delhi")["temp"] == "22°C"
```

#### 🚨 **Risks of Monkey Patching (with Real-Life Example)**

#### 🎯 **Scenario: Monkey Patching `datetime.now()` in a large codebase**

Let’s say you're working on a large finance application, and for testing purposes, you **monkey patch `datetime.datetime.now()`** to return a fixed time (e.g., for simulating year-end billing).

#### 🔧 Your Patch:
```python
import datetime

class FixedDateTime(datetime.datetime):
    @classmethod
    def now(cls):
        return cls(2024, 12, 31, 23, 59)

datetime.datetime = FixedDateTime  # ← Monkey patching!
```

#### ⚠️ **What Could Go Wrong?**

#### ❌ 1. **Affects Global Behavior Unexpectedly**
You might have **logging**, **token expiration**, **cache keys**, or **audit trail** features across your app — all relying on `datetime.now()`. Now **everything thinks it’s New Year's Eve 2024**, leading to:

- Expired tokens not expiring
- Bills calculated for the wrong period
- Logs becoming untrustworthy
- Caches never invalidating

#### ❌ 2. **Breaks Third-party Libraries**
Imagine a library you're using — say for OAuth tokens — relies on `datetime.now()` to check if a token is still valid.

Now it **fails silently or behaves incorrectly**, because `datetime.now()` is no longer returning the real time.

#### ❌ 3. **Threading Issues & Unpredictable Behavior**
If your patch happens **only in one module** or **one thread**, and another part of the code (or thread) expects the original `datetime`, it creates **race conditions** and **hard-to-debug bugs**.

#### ❌ 4. **Debugging Hell**
Later, someone else working on the code gets this error:

```python
TypeError: can't instantiate datetime.datetime
```

Turns out your monkey patch removed core functionality or modified behavior deeply — and no one knows where or why, because it’s hidden in a utility module.

#### 🧠 **Interview Insight: How to Handle It Better**

If you need to override functionality like this:
- ✅ Use `unittest.mock.patch()` in test cases
- ✅ Use **dependency injection** (pass in a function or time provider)
- ✅ Keep monkey patches **isolated** and **documented** (never silent or hidden)
- ❌ Never monkey patch globally in production code

#### 🔒 **Summary of Risks:**

| Risk                            | Description |
|---------------------------------|-------------|
| **Unintended Side Effects**     | Affects unrelated code |
| **Breaks Library Internals**    | Many libraries rely on standard behavior |
| **Difficult to Trace Bugs**     | Changes aren't obvious |
| **Race Conditions**             | Different threads/modules behave inconsistently |
| **Testing vs Production Conflicts** | Patch left behind causes production bugs |

### Q3. What is Auto-Registering? <Just read the definition>
> Auto-registering means that **as soon as a class is defined**, it's automatically added to a **registry or lookup table** for later use.

#### 🧠 Why Use It?

- Avoids repetitive manual registration
- Useful in plug-in architectures (e.g., new `Commands`, `Tasks`, `Serializers`)
- Keeps things decoupled and modular

#### 🧪 Real-Life Example: Command Registry

#### 🎯 Goal:
When someone defines a new `Command` subclass, it should automatically be added to a global `registry`.

#### 🔧 Step-by-step with a Metaclass

```python
# Global registry
command_registry = {}

# Custom Metaclass
class AutoRegisterMeta(type):
    def __new__(cls, name, bases, dct):
        new_class = super().__new__(cls, name, bases, dct)
        if name != 'BaseCommand':
            command_registry[name] = new_class
        return new_class

# Base class using the metaclass
class BaseCommand(metaclass=AutoRegisterMeta):
    pass

# Now any subclass will be auto-registered
class HelloCommand(BaseCommand):
    def run(self):
        print("Hello!")

class GoodbyeCommand(BaseCommand):
    def run(self):
        print("Goodbye!")

# The registry now has all command classes!
print(command_registry)
```

#### ✅ Output:
```python
{
 'HelloCommand': <class '__main__.HelloCommand'>,
 'GoodbyeCommand': <class '__main__.GoodbyeCommand'>
}
```

#### 🔥 Bonus: Dynamic Execution from Registry

```python
command_name = 'HelloCommand'
cmd_class = command_registry[command_name]
cmd_instance = cmd_class()
cmd_instance.run()  # Output: Hello!
```

#### 🧠 How to Explain in an Interview

> “Auto-registering via metaclasses is a clean pattern for managing class discovery in large or extensible systems. It decouples class creation from usage, and allows dynamic behavior like plugin loading or class-based routing without manually keeping a registry.”

#### 🔐 Summary

| Feature                | Benefit |
|------------------------|---------|
| Automatic discovery    | No manual registration needed |
| Centralized registry   | Easily track all subclasses |
| Scalable & modular     | Great for plugins & extensibility |
| Cleaner architecture   | No if/else chains or boilerplate |
