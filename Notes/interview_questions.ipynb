## 🔹 Section 1: Core Python Concepts

### Q1: What is the difference between `_var`, `__var`, and `__var__` in Python?

**Answer:**
- `_var`: A convention to indicate the variable is intended for internal use - protected variable - accessed with class and subclass. Although Python doesn’t prevent you from accessing it directly outside the class, doing so goes against the intended usage.
- `__var`: Triggers name mangling to prevent accidental access in subclasses. Internally renamed to `_ClassName__var` - private vars - accessed within class
- `__var__`: Reserved for special methods like `__init__`, `__str__`, etc. Avoid defining your own `__var__` names unless implementing dunder methods.
- Ref.: [Reddit discussion about vars](https://www.reddit.com/r/learnpython/comments/k0rhy8/when_to_use_private_variables_in_python/)

**Scenario:**
You're building a base class that contains a sensitive method that shouldn't be overridden or accessed by subclasses. Using `__sensitive_method` protects it from accidental override.

### Q2: Explain `*args` and `**kwargs` with an example.

**Answer:**
- `*args` allows passing a variable number of **positional** arguments. *Unpacked* as tuple
- `**kwargs` allows passing a variable number of **keyword** arguments. *Unpacked* as dictionary

```python
def example_func(*args, **kwargs):
    print("Args:", args)
    print("Kwargs:", kwargs)

example_func(1, 2, 3, a=10, b=20)
```

**Output:**
```
Args: (1, 2, 3)
Kwargs: {'a': 10, 'b': 20}
```

**Scenario:**
In a data processing pipeline, you may have dynamic filters or parameters passed at runtime. `*args` and `**kwargs` help keep the interface flexible and extensible.

**Ref.:**
- https://www.programiz.com/python-programming/args-and-kwargs

### Q3: What are closures in Python?

**Answer:**
A closure is a function object that remembers values in enclosing scopes even if they are not in memory anymore.

```python
def outer(x):
    def inner(y):
        return x + y
    return inner

closure_fn = outer(10)
print(closure_fn(5))  # Output: 15
```

**Scenario:**
You’re building a data transformation engine where you want each stage to carry over config values like thresholds. Closures help carry the config context across function calls.

**Ref.:**
- [Closures](https://realpython.com/python-closure/#:~:text=In%20Python%2C%20a%20closure%20is,feature%20in%20functional%20programming%20languages.)

### Q4: What is a decorator in Python? How do you use it practically?

**Answer:**
A decorator is a higher-order function that wraps another function to modify its behavior without changing its code.

```python
def log_decorator(func):
    def wrapper(*args, **kwargs):
        print(f"Calling {func.__name__}")
        return func(*args, **kwargs)
    return wrapper

@log_decorator
def greet(name):
    return f"Hello {name}"

print(greet("Alice"))
```

**Scenario:**
You're implementing access control in a web app. You could use `@login_required` to guard protected views.

**Ref:**
- https://www.programiz.com/python-programming/decorator

### Q5: What are iterators in Python?

**Answer:**
An iterator is an object which implements the `__iter__()` and `__next__()` methods and can be used to loop through data.

```python
nums = [1, 2, 3]
iter_obj = iter(nums)
print(next(iter_obj))  # Output: 1
```

### Q6: What are generators? How are they different from iterators?

**Answer:**
Generators are a simpler way to create iterators using `yield`.

```python
def countdown(n):
    while n > 0:
        yield n
        n -= 1
```

**Scenario:**

### Q7: How does exception handling work in Python?

**Answer:**
Exception handling in Python is done using `try`, `except`, `else`, and `finally` blocks:

```python
try:
    result = 10 / 0
except ZeroDivisionError:
    print("Can't divide by zero")
else:
    print("No error occurred")
finally:
    print("This block always runs")
```

- `try`: Wraps code that may raise an exception
- `except`: Catches and handles specific exceptions
- `else`: Runs if no exception is raised
- `finally`: Always executes (for cleanup or logging)

**Scenario:**
You're calling a third-party API and want to handle connection timeouts gracefully, log the issue, and retry if needed.

### Q8: Difference between `for` and `while` loops in Python?

**Answer:**
- `for` loops are used when you know ahead of time how many iterations you want.
- `while` loops are used when the iteration depends on a condition rather than a range or list.

```python
for i in range(3):
    print(i)  # prints 0,1,2

x = 0
while x < 3:
    print(x)
    x += 1
```

**Scenario:**
Use a `for` loop to process items in a list. Use a `while` loop to poll an API every few seconds until you get a success response.

## 🔹 Section 2: Object-Oriented Programming

### Q1: What is the difference between a class, metaclass, and abstract class in Python?

**Answer:**
- **Class**: Blueprint for creating objects (instances).
- **Metaclass**: Class of a class; controls class creation behavior.
- **Abstract Class**: A class that can't be instantiated directly and must be subclassed.

```python
from abc import ABC, abstractmethod

class Animal(ABC):
    @abstractmethod
    def sound(self):
        pass

class Dog(Animal):
    def sound(self):
        return "Bark"
```

**Scenario:**
You’re designing a plugin architecture. Abstract classes enforce a contract. Metaclasses can be used to auto-register plugins.

### Q2: Explain inheritance, method overriding, and overloading in Python.

**Answer:**
- **Inheritance**: One class derives from another.
- **Overriding**: Subclass redefines a method of the parent.
- **Overloading**: Python doesn’t support traditional method overloading; simulate using default/`*args`.

```python
class Parent:
    def speak(self):
        return "Parent speaking"

class Child(Parent):
    def speak(self):  # overriding
        return "Child speaking"
```

**Scenario:**
Override `__str__()` in child class to customize printed object info for logging.

### Q3: What are dunder methods? Name a few useful ones.

**Answer:**
Dunder (double underscore) methods allow operator overloading and integration with Python internals.

Common ones:
- `__init__`, `__str__`, `__repr__`, `__len__`, `__eq__`, `__getitem__`, `__setitem__`

```python
class Point:
    def __init__(self, x, y):
        self.x = x
        self.y = y

    def __str__(self):
        return f"({self.x}, {self.y})"
```

**Scenario:**
Use `__eq__` in a data model class to enable easy object comparison in test cases.

### Q7: How do you create a custom iterator?

```python
class Countdown:
    def __init__(self, start):
        self.start = start

    def __iter__(self):
        return self

    def __next__(self):
        if self.start <= 0:
            raise StopIteration
        val = self.start
        self.start -= 1
        return val

for num in Countdown(5):
    print(num)
```

### Q10: How do you write your own context manager?

```python
class FileOpener:
    def __init__(self, filename, mode):
        self.file = open(filename, mode)

    def __enter__(self):
        return self.file

    def __exit__(self, exc_type, exc_val, exc_tb):
        self.file.close()

with FileOpener('test.txt', 'w') as f:
    f.write("hello")
```

### Q4: Explain defaultdict, OrderedDict, and frozenset.

```python
from collections import defaultdict, OrderedDict

# defaultdict
dd = defaultdict(int)
dd['a'] += 1
print(dd['a'])  # 1

# OrderedDict
o = OrderedDict()
o['a'] = 1
o['b'] = 2
print(list(o.keys()))  # ['a', 'b']

# frozenset
fs = frozenset([1, 2, 2, 3])
print(fs)  # frozenset({1, 2, 3})
```

**Scenario:** defaultdict simplifies counter-style patterns, OrderedDict preserves order (before Python 3.7 dict did not), frozenset is a hashable set.

## 🔹 Section 5: Concurrency and Parallelism

### Q1: What is the difference between threading and multiprocessing in Python?

**Answer:**
- **Threading**: Runs multiple threads in a single process. Threads share memory space but are limited by the GIL (Global Interpreter Lock).
- **Multiprocessing**: Spawns separate processes with their own Python interpreter and memory space, bypassing GIL.

```python
from threading import Thread
import multiprocessing

# Threading example
def print_nums():
    for i in range(3):
        print(f"Thread {i}")

Thread(target=print_nums).start()

# Multiprocessing example
def print_proc():
    print("From process")

if __name__ == '__main__':
    multiprocessing.Process(target=print_proc).start()
```

**Scenario:** Use multiprocessing for CPU-bound tasks; threading for I/O-bound tasks.

---

### Q2: What are race conditions and how can you prevent them?

**Answer:** A race condition occurs when threads/processes access shared data simultaneously and the outcome depends on the order of execution.

**Prevention Tools:** Locks, Semaphores, Queues

```python
import threading

lock = threading.Lock()
shared_counter = 0

def increment():
    global shared_counter
    with lock:
        for _ in range(1000):
            shared_counter += 1

threads = [threading.Thread(target=increment) for _ in range(5)]
[thread.start() for thread in threads]
[thread.join() for thread in threads]
print(shared_counter)  # Expected: 5000
```

### Q5: What is pooling in concurrency?

**Answer:** Pooling is a technique to reuse a fixed number of threads/processes to execute many tasks efficiently.

```python
from multiprocessing import Pool

def square(x): return x * x

with Pool(4) as p:
    print(p.map(square, [1, 2, 3, 4]))  # Output: [1, 4, 9, 16]
```

### Q7: What is Method Overriding and Overloading in Python?

**Overriding:** Redefining a parent class method in a child class.

```python
class Parent:
    def greet(self):
        print("Hello from Parent")

class Child(Parent):
    def greet(self):
        print("Hello from Child")

Child().greet()  # Output: Hello from Child
```

**Overloading:** Python doesn’t support traditional method overloading. Achieved via default args or `*args`, `**kwargs`.

```python
class Example:
    def greet(self, name=None):
        if name:
            print(f"Hello {name}")
        else:
            print("Hello")

Example().greet("Alice")  # Output: Hello Alice
```
