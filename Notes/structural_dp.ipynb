### **Decorator Design Pattern**  
The **Decorator Pattern** is a **structural design pattern** that allows adding **new functionalities dynamically** to an object **without modifying its structure**. It’s useful when you want to **extend** an object’s behavior **at runtime**.

---

### **🔹 When to Use the Decorator Pattern?**  
✅ When you need to **add behavior dynamically** without altering the original class.  
✅ When you want to follow the **Open/Closed Principle** → **Open for extension, closed for modification**.  
✅ When **subclassing** leads to an explosion of classes (**too many subclasses**).  

---

### **🔹 Example: Coffee Shop Customization**  

#### **Scenario:**  
You are designing a coffee shop system where customers can order coffee with **add-ons like milk, sugar, or caramel**. Instead of modifying the base `Coffee` class, you can **decorate it dynamically** with different add-ons.

---

### **🔹 Key Advantages of the Decorator Pattern**
✅ **Extensibility** → New behaviors can be added **without modifying** existing code.  
✅ **Composability** → Multiple decorators can be combined flexibly.  
✅ **Single Responsibility Principle** → Separates concerns, avoiding large monolithic classes.  

---

### **🔹 Gotchas / Disadvantages**
⚠ **Complexity Increases** → Too many decorators can make code harder to read.  
⚠ **Order of Wrapping Matters** → If decorators depend on each other, the sequence matters.  
⚠ **Debugging Can Be Harder** → Since behavior is dynamically wrapped, it may be harder to trace execution.  

---

### **🚀 Summary**
- The **Decorator Pattern** dynamically extends an object’s behavior **without modifying the base class**.  
- Uses **composition** instead of inheritance → More **flexible & scalable**.  
- Great for **feature toggles, UI elements, or adding behaviors dynamically**.  


### **How Does the Order of Wrapping Matter in the Decorator Pattern?**  

The **order of wrapping matters** in the **Decorator Pattern** because **each decorator modifies the output of the previous one**. Since decorators are applied **sequentially**, changing the order can lead to **different behavior and results**.

---

### **🔹 Example: Coffee Decorators with Different Orders**  

Let’s consider the **coffee customization** example again. We'll compare different wrapping orders.

#### **Implementation**
```python
coffee = SimpleCoffee()  # Base coffee
coffee = CaramelDecorator(MilkDecorator(SugarDecorator(coffee)))
print(f"{coffee.description()} - ₹{coffee.cost()}")
```
#### **Output**
```
Simple Coffee, Sugar, Milk, Caramel - ₹85
```

🔹 **Breakdown of Execution Order**:
1. **SugarDecorator** is applied first → Adds "Sugar" & increases cost.
2. **MilkDecorator** is applied next → Adds "Milk" & increases cost.
3. **CaramelDecorator** is applied last → Adds "Caramel" & increases cost.

---

### **🔹 What Happens If We Change the Order?**
Now, let’s **wrap in a different order**:

```python
coffee = SimpleCoffee()
coffee = SugarDecorator(CaramelDecorator(MilkDecorator(coffee)))
print(f"{coffee.description()} - ₹{coffee.cost()}")
```
#### **Output**
```
Simple Coffee, Milk, Caramel, Sugar - ₹85
```
🔹 **Notice the difference?**  
- The **cost remains the same**, but the **description order is different**.  
- This can matter if some decorators depend on the order in which they are applied.

---

### **🔹 When Does Order Affect Functionality?**
1. **Logging & Security in APIs**  
   ```python
   request = Request()
   request = AuthenticationDecorator(LoggingDecorator(request))
   ```
   - If you **log first**, failed authentication attempts are recorded.
   - If you **authenticate first**, failed attempts may not be logged.

2. **Text Formatting Example**  
   ```python
   text = PlainText("Hello")
   text = BoldDecorator(ItalicDecorator(text))
   print(text.render())
   ```
   **vs.**
   ```python
   text = ItalicDecorator(BoldDecorator(PlainText("Hello")))
   ```
   - **Bold first, then Italic** → `<b><i>Hello</i></b>`
   - **Italic first, then Bold** → `<i><b>Hello</b></i>`

3. **Middleware in Web Frameworks (e.g., Flask, Django)**  
   - If **caching comes before authentication**, users might see **cached** unauthorized responses.
   - If **authentication comes first**, only valid users can access cached content.

---

### **🔹 Conclusion**
✅ **If decorators are independent**, order **does not matter** (e.g., adding sugar vs. milk).  
✅ **If decorators depend on each other**, order **changes behavior** (e.g., security, middleware, formatting).  


### **Proxy Design Pattern**  
The **Proxy Pattern** is a **structural design pattern** that provides a **placeholder or surrogate** for another object to **control access to it**.  

---

### **🔹 When to Use the Proxy Pattern?**  
✅ **Access Control** → Restrict access to sensitive objects (**authentication, authorization**).  
✅ **Lazy Initialization** → Load expensive objects **only when needed** (**virtual proxy**).  
✅ **Logging/Monitoring** → Track method calls (**logging proxy**).  
✅ **Caching** → Store results to improve performance (**caching proxy**).  
✅ **Security & Protection** → Prevent unauthorized actions (**protection proxy**).  

---

## **🔹 Types of Proxy Patterns**
1. **Virtual Proxy** → Creates expensive objects **on demand**.  
2. **Protection Proxy** → Controls access to an object based on permissions.  
3. **Remote Proxy** → Represents an object **in a different location (e.g., API, database, network requests)**.  
4. **Smart Proxy** → Adds extra functionality (e.g., logging, caching, request tracking).  

---

### **🔹 Other Proxy Pattern Use Cases**
- **Logging Proxy** → Logs method calls for debugging.  
- **Caching Proxy** → Stores results to prevent repeated computations.  
- **Virtual Proxy** → Loads large objects **only when needed** (e.g., image loading in GUI apps).  
- **Remote Proxy** → Represents objects from **remote servers (e.g., APIs, database queries, microservices, RPCs)**.  

---

### **🔹 Key Advantages of Proxy Pattern**
✅ **Security** → Controls access to sensitive operations.  
✅ **Performance Optimization** → Uses **lazy loading & caching** for expensive objects.  
✅ **Separation of Concerns** → Keeps the real object **clean** while handling extra functionality in the proxy.  

---

### **🔹 Gotchas / Disadvantages**
⚠ **Extra Complexity** → Adds more classes, increasing maintenance.  
⚠ **Performance Overhead** → Proxy may slow down operations if not optimized.  
⚠ **Debugging Challenges** → Harder to trace issues as requests go through **intermediary objects**.  

---

### **🚀 Summary**
- The **Proxy Pattern** acts as a **middleman** to control access to an object.  
- Used for **security, caching, logging, lazy loading, and access control**.  
- Can be applied in **database access, API calls, file handling, and network operations**.
