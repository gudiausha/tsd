## **Shortcomings of `unittest` and Why Choose `pytest`**  

### **🚨 Problems with `unittest`**
Even though `unittest` is Python’s built-in testing framework, it has several **drawbacks** that make `pytest` a better choice.  

| **Shortcoming in `unittest`** | **How `pytest` Fixes It** |
|--------------------------------|--------------------------|
| **Verbose Assertions** - Uses `self.assertEqual(a, b)` instead of plain `assert a == b` | ✅ `pytest` allows direct `assert a == b`, making tests more readable |
| **Test Discovery Requires Boilerplate** - Tests must be inside a class that inherits `unittest.TestCase` | ✅ `pytest` auto-discovers test functions, no need for classes |
| **No Built-in Parameterization** - You need to loop inside the test method | ✅ `pytest.mark.parametrize` makes it easy to test multiple cases |

| **Poor Failure Reports** - Default output is minimal | ✅ `pytest` gives **detailed failure reports**, including actual and expected values |

## **🔥 `pytest` Fixtures - What & Why?**  

### **🔹 What are Fixtures?**
Fixtures in `pytest` are **functions that set up some context** for a test and then **clean up** after the test completes. They help avoid redundant setup/teardown code and make tests more maintainable.  

Instead of using `setUp()` and `tearDown()` (like in `unittest`), `pytest` fixtures provide **flexible and reusable test setup mechanisms**.

---

## **💡 Why Use Fixtures?**
- **Avoid Code Duplication** → No need to write the same setup logic in multiple tests.  
- **Automatic Setup & Cleanup** → You don’t need to call teardown methods manually.  
- **Dependency Injection** → Tests **request** fixtures instead of calling them explicitly.  
- **Scope Control** → Can run **per function, class, module, or session**.  

---

## **🛠 Basic Fixture Example**
🚫 **Without Fixtures (Setup in Every Test)**
```python
import sqlite3

def test_query1():
    conn = sqlite3.connect(":memory:")  # Setup
    cursor = conn.cursor()
    cursor.execute("CREATE TABLE users (id INTEGER, name TEXT)")
    conn.close()  # Teardown

def test_query2():
    conn = sqlite3.connect(":memory:")  # Setup again
    cursor = conn.cursor()
    cursor.execute("CREATE TABLE users (id INTEGER, name TEXT)")
    conn.close()  # Teardown again
```

✅ **With Fixtures (Reusable Setup)**
```python
import pytest
import sqlite3

@pytest.fixture
def db():
    """Fixture to create and return a database connection."""
    conn = sqlite3.connect(":memory:")  # Setup
    yield conn  # Everything before yield is setup, after yield is teardown
    conn.close()  # Teardown

def test_query1(db):
    cursor = db.cursor()
    cursor.execute("CREATE TABLE users (id INTEGER, name TEXT)")
    assert cursor is not None

def test_query2(db):
    cursor = db.cursor()
    cursor.execute("CREATE TABLE products (id INTEGER, name TEXT)")
    assert cursor is not None
```
🎯 **Why is this better?**
- The `db()` fixture **runs automatically** when a test requests `db`.  
- The **setup (DB connection) runs before the test**, and **teardown (closing the DB) runs after**.  
- **No duplicate setup code** in multiple tests.  

---

## **⚡️ Fixture Scope: Function, Class, Module, Session**
By default, fixtures run **for each test function**. You can change this using `@pytest.fixture(scope="...")`.

| **Scope**   | **Runs...** |
|------------|------------|
| `function` | **(default)** Runs once per test function |
| `class`    | Runs once per test class |
| `module`   | Runs once per module (single `.py` file) |
| `session`  | Runs once per entire test session |

**Example: A fixture that runs only once per test module**
```python
@pytest.fixture(scope="module")
def db():
    conn = sqlite3.connect(":memory:")  
    yield conn  
    conn.close()  
```
🚀 **Saves time** when setting up expensive resources!

---

## **🔄 Using Fixtures with `autouse` (Implicit Execution)**
Fixtures normally require tests to **explicitly request them**, but you can make them **run automatically** using `autouse=True`.

✅ **Example: Automatically Logging Every Test**
```python
@pytest.fixture(autouse=True)
def log():
    print("\n🔹 Test Started")
    yield
    print("✅ Test Finished")
```
Now, **all tests** will print logs without needing to request `log()`.

---

## **🔁 Parametrized Fixtures (Reusable Setup Variants)**
Instead of writing **separate fixtures for different setups**, `pytest` allows **parameterized fixtures**.

✅ **Example: Testing with Different User Roles**
```python
@pytest.fixture(params=["admin", "user", "guest"])
def user_role(request):
    return request.param  # Returns "admin", "user", or "guest"

def test_access_control(user_role):
    assert user_role in ["admin", "user", "guest"]
```
🎯 **Why is this useful?**
- Runs **three different tests**: One for `"admin"`, one for `"user"`, and one for `"guest"`.  
- **No need to write three separate tests**!  

---

## **🕵️‍♂️ Mocking with Fixtures**
Fixtures are **great for mocking** dependencies in tests.

✅ **Example: Mocking an API Request**
```python
import pytest

class APIClient:
    def get_data(self):
        return "Real Data"

@pytest.fixture
def mock_api(mocker):
    mock = mocker.patch.object(APIClient, "get_data", return_value="Mock Data")
    return mock

def test_api(mock_api):
    client = APIClient()
    assert client.get_data() == "Mock Data"
```
🚀 **Why use this?**
- **Replaces `get_data()` with a mock returning "Mock Data"**.  
- **Prevents actual API calls**, making tests **faster and isolated**.  

---

## **🚀 Key Takeaways**
✅ **Fixtures = Reusable Setup + Automatic Cleanup**  
✅ **Avoid Duplicate Code** (No need for `setUp()/tearDown()`)  
✅ **Flexible Scoping** (Per function, class, module, or session)  
✅ **Auto-run Fixtures** (`autouse=True` runs them for all tests)  
✅ **Parameterization** (`params=[...]` runs a fixture multiple times with different values)  
✅ **Great for Mocking** (Use `pytest-mock` for replacing real dependencies)  


## **🐒 Monkey Patching in Python (with pytest)**  

### **🔹 What is Monkey Patching?**
Monkey patching is the **dynamic modification of a class, module, or function at runtime**. It allows you to **override or replace behavior without modifying the original source code**.

🚀 **Why Use It?**  
- **Mock dependencies in tests**  
- **Change the behavior of third-party libraries**  
- **Quickly patch bugs** in legacy code  

---

## **💡 Example 1: Monkey Patching a Function**
Imagine we have a module `math_ops.py`:

```python
# math_ops.py
def add(a, b):
    return a + b
```

Now, let’s monkey patch it **during runtime**:

```python
import math_ops

def fake_add(a, b):
    return 42  # Always returns 42

math_ops.add = fake_add  # Monkey patching the original function

print(math_ops.add(2, 3))  # Output: 42
```

🔹 **We replaced `add(a, b)` with `fake_add(a, b)`, which always returns `42`!**  

---

## **🐍 Monkey Patching in Tests (with `pytest-mock`)**  

✅ **Example 2: Mocking a Method in a Class**  
Let’s say we have this class:

```python
# api.py
class APIClient:
    def fetch_data(self):
        return "Real Data"
```

We want to **replace** `fetch_data()` during testing so it doesn’t make a real API call.

```python
import pytest
from api import APIClient

def fake_fetch():
    return "Mock Data"

def test_api():
    APIClient.fetch_data = fake_fetch  # Monkey patching the method
    client = APIClient()
    assert client.fetch_data() == "Mock Data"  # ✅ Test passes!
```

---

## **🐍 Monkey Patching with `pytest-mock` (Recommended)**
Instead of **manually replacing methods**, use `pytest-mock` for **safe monkey patching**.

✅ **Example 3: Using `mocker.patch.object()`**
```python
def test_api(mocker):
    mocker.patch.object(APIClient, "fetch_data", return_value="Mock Data")
    
    client = APIClient()
    assert client.fetch_data() == "Mock Data"  # ✅ Uses the patched version!
```
🎯 **Why is this better?**
- **Automatic restoration** after the test
- **Cleaner and safer** than manual patching

---

## **🚨 Monkey Patching Gotchas**
1️⃣ **⚠️ Affects Global State** → If you monkey patch a module, **it applies everywhere**  
2️⃣ **⚠️ Hard to Debug** → Unexpected changes may cause issues in other parts of the code  
3️⃣ **⚠️ Not Always Thread-Safe** → Be careful in **multithreading scenarios**  

---

## **🚀 Key Takeaways**
✅ **Monkey patching dynamically replaces behavior at runtime**  
✅ **Use it for testing, mocking dependencies, or patching third-party code**  
✅ **Use `pytest-mock` for safe and temporary patching in tests**  
✅ **Be cautious, as it modifies global state**  


## **🎯 What Are Marks in `pytest`?**  

In `pytest`, **marks (or markers)** are **special decorators** used to categorize, filter, or modify test behavior.  

They help in:  
✅ **Grouping & Selecting Tests** (e.g., run only slow tests)  
✅ **Skipping or Expecting Failures** (e.g., skip tests based on conditions)  
✅ **Parameterizing Tests** (e.g., run a test with multiple inputs)  

---

## **🔥 Common Types of Marks**
### **1️⃣ Skipping Tests (`@pytest.mark.skip`)**
🚀 Use this when a test is **not relevant or temporarily disabled**.

```python
import pytest

@pytest.mark.skip(reason="Feature not implemented yet")
def test_new_feature():
    assert 1 + 1 == 2  # This test will be skipped
```

---

### **2️⃣ Conditionally Skipping Tests (`@pytest.mark.skipif`)**
Skip tests **based on a condition** (e.g., Python version, OS, environment variables).

```python
import pytest
import sys

@pytest.mark.skipif(sys.version_info < (3, 8), reason="Requires Python 3.8+")
def test_new_feature():
    assert 1 + 1 == 2  # Skipped if Python < 3.8
```

---

### **3️⃣ Expected Failures (`@pytest.mark.xfail`)**
Use this when a test is **expected to fail**, but you don’t want the test suite to fail.

```python
@pytest.mark.xfail(reason="Known bug in production")
def test_buggy_function():
    assert 1 / 0  # This will fail, but pytest won’t count it as a failure
```

---

### **4️⃣ Running Slow Tests Separately (`@pytest.mark.slow`)**
You can create custom markers (e.g., for slow tests) and **run them selectively**.

```python
@pytest.mark.slow
def test_large_dataset():
    assert sum(range(1000000)) > 0
```
Run only slow tests:  
```sh
pytest -m slow
```

---

### **5️⃣ Running Tests with Multiple Inputs (`@pytest.mark.parametrize`)**
Instead of writing multiple tests, `@pytest.mark.parametrize` allows you to **run a test with different inputs**.

```python
import pytest

@pytest.mark.parametrize("x, y, expected", [
    (1, 2, 3),
    (4, 5, 9),
    (-1, -1, -2)
])
def test_addition(x, y, expected):
    assert x + y == expected
```
**This runs three tests automatically!** 🎯

---

### **6️⃣ Custom Markers (`pytest.ini` Configuration)**
If you define your own markers, register them in `pytest.ini` to avoid warnings.

```ini
# pytest.ini
[pytest]
markers =
    slow: marks tests as slow
    db: marks tests that use a database
```

Now you can use them:
```python
@pytest.mark.db
def test_database_query():
    assert fetch_data() is not None
```
Run only DB tests:
```sh
pytest -m db
```

---

## **🚀 Key Takeaways**
✅ **Marks help organize, skip, and parameterize tests**  
✅ **Use `@pytest.mark.skip` and `@pytest.mark.skipif` to skip tests**  
✅ **Use `@pytest.mark.parametrize` for running tests with multiple inputs**  
✅ **Use `@pytest.ini` to register custom markers**  


# **🔥 1. Fixtures (`@pytest.fixture`)**
### **📌 What It Is:**  
- **Fixtures** handle **setup and teardown** logic in tests (e.g., opening/closing database connections).  
- Reduces **code duplication** across multiple tests.  

### **✅ Example:**
```python
import pytest

@pytest.fixture
def sample_data():
    return {"name": "Alice", "age": 25}

def test_data(sample_data):
    assert sample_data["name"] == "Alice"
```
📌 **The fixture `sample_data()` is automatically injected into `test_data()`**.

### **🚨 Gotcha:**  
- Fixtures **execute once per test function by default**, but you can **scope them**:
  ```python
  @pytest.fixture(scope="module")  # Runs once per module
  def db_connection():
      return create_db_connection()
  ```

# **🔥 2. Parametrization (`@pytest.mark.parametrize`)**
### **📌 What It Is:**  
- Runs **the same test with different inputs** (avoids writing redundant tests).  

### **✅ Example:**
```python
import pytest

@pytest.mark.parametrize("x, y, expected", [(1, 2, 3), (4, 5, 9), (-1, -1, -2)])
def test_addition(x, y, expected):
    assert x + y == expected
```
📌 **Runs three tests automatically with different values**.

### **🚨 Gotcha:**  
- If using **parametrize with fixtures**, use `indirect=True`:
  ```python
  @pytest.mark.parametrize("setup_data", [10, 20], indirect=True)
  def test_with_fixture(setup_data):
      assert setup_data in [10, 20]
  ```

### **📌 What It Is:**  
- **Replaces dependencies** (e.g., mock API calls, database interactions).  

### **✅ Example: Using `monkeypatch`**
```python
import pytest

def get_env():
    return "prod"

def test_mock_env(monkeypatch):
    monkeypatch.setattr("get_env", lambda: "test")
    assert get_env() == "test"
```

## **8️⃣ Test-Driven Development (TDD) with `pytest`**
**TDD Flow**  
1. **Write a failing test first**  
2. **Write the minimal code to pass the test**  
3. **Refactor the code while keeping tests green**  

### **Example**
```python
# Step 1: Write the failing test first (Function not yet implemented)
def test_add():
    assert add(2, 3) == 5
```

```python
# Step 2: Implement minimal function
def add(a, b):
    return a + b  # Now test passes
```

✅ **Why TDD?**  
- Forces **modular design**.  
- Reduces **debugging time** later.

## **9️⃣ Integrating `pytest` with CI/CD**
### **Why Automate Tests in CI/CD?**
- Ensures all tests **run on every commit**.  
- Prevents merging **bad code** into production.  

### **Example: Running `pytest` in GitHub Actions**
```yaml
name: Run Pytest
on: [push, pull_request]
jobs:
  test:
    runs-on: ubuntu-latest
    steps:
      - uses: actions/checkout@v3
      - name: Set up Python
        uses: actions/setup-python@v3
        with:
          python-version: "3.8"
      - name: Install dependencies
        run: pip install -r requirements.txt
      - name: Run tests
        run: pytest --cov=app
```

✅ **How to Integrate with Other CI/CD Tools?**  
- **Jenkins**: Use `pytest` inside pipeline scripts.  
- **GitLab CI/CD**: Use `pytest` in `.gitlab-ci.yml`.
