**Interviewer:** *What is the difference between multiprocessing and multithreading?*

**Me (Interviewee):**  
Sure! The primary difference between multiprocessing and multithreading lies in how they achieve concurrency and how system resources are utilized.

- **Multithreading** involves multiple threads running within the same process. These threads share the same memory space and resources, which makes inter-thread communication easier and more efficient. It's generally well-suited for **I/O-bound tasks** — like waiting for user input, reading from files, or making API calls — where threads can remain idle while waiting, allowing others to proceed.

- On the other hand, **multiprocessing** involves spawning multiple **independent processes**, each with its own memory space and Python interpreter. This approach is ideal for **CPU-bound tasks** such as heavy data processing or computations. In Python, multiprocessing overcomes the Global Interpreter Lock (GIL), allowing true parallel execution across multiple CPU cores.

To summarize:
- **Multithreading** is lightweight, shares memory, but is constrained by the GIL in CPython for CPU-bound tasks.
- **Multiprocessing** is heavier in terms of resource usage but enables real parallelism and is the preferred approach for CPU-intensive workloads.


**Interviewer:** *How can we implement polymorphism in Python?*

**Me (Interviewee):**  
Polymorphism in Python refers to the ability to use a unified interface to work with objects of different types. It’s one of the core principles of object-oriented programming and can be implemented in multiple ways:

### 1. **Duck Typing (Informal Polymorphism)**
Python follows duck typing — “if it walks like a duck and quacks like a duck, it’s a duck.”  
This means we don’t need strict inheritance. As long as different classes implement the same method names, they can be used interchangeably.

```python
class Dog:
    def speak(self):
        return "Woof!"

class Cat:
    def speak(self):
        return "Meow!"

def make_sound(animal):
    print(animal.speak())

make_sound(Dog())  # Woof!
make_sound(Cat())  # Meow!
```

### 2. **Inheritance-Based Polymorphism**
We can define a base class and override its methods in derived classes.

```python
class Animal:
    def speak(self):
        raise NotImplementedError

class Dog(Animal):
    def speak(self):
        return "Woof!"

class Cat(Animal):
    def speak(self):
        return "Meow!"
```

Here, we can treat both `Dog` and `Cat` as `Animal` objects polymorphically.

### 3. **Using Abstract Base Classes (ABCs)**
To enforce method implementation in subclasses, we can use the `abc` module.

```python
from abc import ABC, abstractmethod

class Animal(ABC):
    @abstractmethod
    def speak(self):
        pass

class Dog(Animal):
    def speak(self):
        return "Woof!"
```

This approach ensures polymorphism while maintaining structure and discipline in the code.

---

So in summary, Python supports **both static (via inheritance)** and **dynamic (via duck typing)** polymorphism, giving developers a lot of flexibility depending on the use case.

**Me (Interviewee):**  
Polymorphism is one of the fundamental principles of object-oriented programming. The term comes from the Greek words *poly* (many) and *morph* (form), meaning **“many forms.”** In programming, polymorphism allows us to use a common interface to interact with different underlying data types or class implementations.

In Python and in general, polymorphism lets us write **flexible and reusable code** — the same function or method can behave differently depending on the object it is acting upon.

---

### 🔸 Types of Polymorphism

#### ✅ **1. Static Polymorphism (Compile-time Polymorphism)**

- It refers to method overloading or operator overloading — where the method to be executed is determined **at compile time** based on the method signature (number and types of parameters).
- Python doesn’t support traditional method overloading as in Java or C++, but we can achieve similar behavior using default arguments or `*args`, `**kwargs`.

```python
# Emulating method overloading
class Calculator:
    def add(self, a, b=0, c=0):
        return a + b + c

calc = Calculator()
print(calc.add(2))        # 2
print(calc.add(2, 3))     # 5
print(calc.add(2, 3, 4))  # 9
```

#### ✅ **2. Dynamic Polymorphism (Runtime Polymorphism)**

- This is where **method overriding** comes into play. The method that gets called is determined **at runtime**, based on the actual object type, not the reference type.
- Python heavily relies on **dynamic polymorphism** via both inheritance and duck typing.

```python
class Shape:
    def area(self):
        pass

class Circle(Shape):
    def area(self):
        return "πr²"

class Square(Shape):
    def area(self):
        return "side²"

def print_area(shape):
    print(shape.area())

print_area(Circle())  # πr²
print_area(Square())  # side²
```

---

### 🧠 Summary

| Type                 | Static Polymorphism                  | Dynamic Polymorphism                   |
|----------------------|--------------------------------------|----------------------------------------|
| Resolution Time      | Compile time                         | Runtime                                |
| Achieved By          | Method/Operator overloading          | Method overriding / duck typing        |
| Python Support       | Partial (via `*args`, operator overloading) | Full support via inheritance & dynamic typing |

**Interviewer:** *Does Python support method overloading? How is it handled in Java, and how is it different in Python?*

In **Java**, method overloading is **natively supported**. That means you can define multiple methods in the same class with the **same name** but **different parameter types or counts**, and the appropriate method is chosen at **compile-time**.

#### 📌 Java Example:
```java
class MathUtils {
    int add(int a, int b) {
        return a + b;
    }

    double add(double a, double b) {
        return a + b;
    }

    int add(int a, int b, int c) {
        return a + b + c;
    }
}
```
- Here, the method to be invoked is resolved **statically** based on parameter types/counts.

---

### ✅ In contrast, Python **does not support traditional method overloading**.

If you define multiple methods with the same name in a class, the **last definition overrides the previous ones**.

#### ❌ Python Example (Incorrect Expectation):
```python
class MathUtils:
    def add(self, a, b):
        return a + b

    def add(self, a, b, c):
        return a + b + c

# This will throw: TypeError: add() missing 1 required positional argument
print(MathUtils().add(1, 2))
```

Only the last method `add(self, a, b, c)` is retained — the first one is overwritten.

We use:
1. **Default arguments**
2. **`*args` / `**kwargs`**
3. **Type checks (if needed)**

```python
class MathUtils:
    def add(self, *args):
        return sum(args)

calc = MathUtils()
print(calc.add(1, 2))       # 3
print(calc.add(1, 2, 3))    # 6
```

If needed, we can even do type-based logic:
```python
def add(self, a, b):
    if isinstance(a, str) or isinstance(b, str):
        return str(a) + str(b)
    return a + b
```

**Interviewer:** *Can you explain operator overloading in Python?*

**Me (Interviewee):**  
Yes, definitely!

**Operator overloading** is a feature in Python that allows us to **change the behavior of standard operators** like `+`, `-`, `==`, etc., for **user-defined classes**.

In Python, this is done by defining **special methods** (also known as **magic methods** or **dunder methods**) inside our class. These methods start and end with double underscores, like `__add__`, `__eq__`, `__str__`, etc.

---

### ✅ Real-world example: `+` Operator Overloading

By default, `+` is used for numbers and strings. But we can redefine it for custom objects.

```python
class Point:
    def __init__(self, x, y):
        self.x = x
        self.y = y

    def __add__(self, other):
        return Point(self.x + other.x, self.y + other.y)

    def __str__(self):
        return f"({self.x}, {self.y})"

p1 = Point(1, 2)
p2 = Point(3, 4)
result = p1 + p2   # Internally calls p1.__add__(p2)

print(result)  # Output: (4, 6)
```

Here, we’ve **overloaded the `+` operator** to add two `Point` objects by their coordinates.

---

### 🔧 Common magic methods used in operator overloading:

| Operator | Magic Method   | Description               |
|----------|----------------|---------------------------|
| `+`      | `__add__`      | Addition                  |
| `-`      | `__sub__`      | Subtraction               |
| `*`      | `__mul__`      | Multiplication            |
| `==`     | `__eq__`       | Equality comparison       |
| `<`      | `__lt__`       | Less than                 |
| `str()`  | `__str__`      | String representation     |
| `repr()` | `__repr__`     | Debug representation       |

---

### ⚠️ When should we use it?

- When we want to make our custom classes feel more **natural or intuitive** to use.
- For example, implementing a `Matrix`, `Vector`, or `Money` class where arithmetic operations make sense.

---

### ✅ Summary:

- Operator overloading lets us redefine how Python operators behave for custom objects.
- It improves **readability** and enables more **Pythonic** usage of our classes.
- It's implemented using **magic methods** like `__add__`, `__eq__`, etc.

**Interviewer:** *What are fixtures in pytest?*

**Me (Interviewee):**  
In `pytest`, **fixtures** are a way to **set up and tear down test preconditions** in a reusable and modular way. They are ideal for preparing things like database connections, file systems, or any objects that are required across multiple test functions.

Fixtures help in **avoiding repetitive code** and promote **clean, maintainable test design**.

---

### ✅ Key Features of `pytest` fixtures:
- Declared using the `@pytest.fixture` decorator
- Can have **scopes** (`function`, `class`, `module`, `session`)
- Can be **parameterized**
- Can **depend on other fixtures**

---

### 🧱 Basic Example:

```python
import pytest

# A simple fixture
@pytest.fixture
def sample_data():
    return [1, 2, 3]

def test_sum(sample_data):
    assert sum(sample_data) == 6
```

- Here, `sample_data` is a fixture that returns a list.
- The test function `test_sum` **uses** this fixture by **declaring it as an argument**.

---

### 🔄 Fixture Lifecycle and Scope:

You can control how often the fixture runs using the `scope` parameter:

```python
@pytest.fixture(scope="module")  # Can also be: "function" (default), "class", "session"
def db_connection():
    print("Setting up DB connection")
    yield "DB connection"
    print("Tearing down DB connection")
```

- `yield` lets you run **setup before** and **teardown after** the test.
- If the scope is `"module"`, the setup runs once per module (all tests share it).

---

### 🔁 Parameterized Fixture:

```python
@pytest.fixture(params=[1, 2, 3])
def number(request):
    return request.param

def test_multiples(number):
    assert number * 2 in [2, 4, 6]
```

- This test will run **three times**, once for each parameter.

---

|------------------|-----------------------------------------------------------------------------|
| Reusability      | Avoid duplication in test setups                                            |
| Scope            | Control how often a fixture is run (`function`, `class`, etc.)             |
| Dependency       | Fixtures can use other fixtures                                             |
| Parameterization | Run a test multiple times with different data                              |
| Teardown         | Use `yield` to add teardown logic after the test completes                 |

**Idempotent methods** in the context of HTTP/REST APIs are those that, when called **multiple times** with the same parameters or data, will always produce the **same result**. In other words, **repeating the operation doesn't change the outcome after the initial request**. 

This is a key principle in RESTful APIs, as it ensures **predictability** and **safety**, especially in distributed systems where requests might be retried due to network issues or server failures.

### ✅ **Examples of Idempotent HTTP Methods:**

1. **GET**:  
   - A `GET` request retrieves data from the server, and calling it multiple times (with the same URL and parameters) will always return the same data (unless the resource changes in the backend).  
   - Example: `GET /users/123` — it will always return the same user data (as long as the data doesn't change).

2. **PUT**:  
   - A `PUT` request is used to **update or replace** a resource. If you send the same `PUT` request multiple times with the same data, the resource will remain the same.  
   - Example: `PUT /users/123` with `{ "name": "John" }` — calling it repeatedly will always set the name to "John", even if it's already set to that.

3. **DELETE**:  
   - A `DELETE` request removes a resource. If the resource doesn't exist, repeated `DELETE` requests will not cause any further effect (they will return a 404 or 204, but no new resource will be deleted).  
   - Example: `DELETE /users/123` — if the user is already deleted, a subsequent call won't affect anything.

4. **PATCH**:  
   - A `PATCH` request partially updates a resource. When the same request is made multiple times with the same parameters, the resource stays the same. This is **idempotent** as long as the update is not altering the state more than once.  
   - Example: `PATCH /users/123` with `{ "email": "new_email@example.com" }` — calling it multiple times won't keep changing the email value.

---

### 🚫 **Non-Idempotent HTTP Methods:**

1. **POST**:  
   - `POST` is **not idempotent**. Each time a `POST` request is made, it typically creates a new resource. Hence, repeating a `POST` request might result in multiple records being created.  
   - Example: `POST /users` — every time you send the request, a new user is created.

---

### 🔄 **Why is Idempotency Important?**

- **Reliability**: It ensures that retries (due to failures like network or server issues) won’t cause unintended side effects.
- **Predictability**: Helps in understanding how APIs behave when the same request is made multiple times.
- **Safety**: Especially important in APIs that deal with sensitive operations, such as financial transactions or data updates.

---

### ✅ **Summary**:
- **Idempotent** methods (GET, PUT, DELETE, PATCH) can be called multiple times with the same parameters without causing unintended side effects.
- **Non-idempotent** methods (POST) might produce different results on each call.


**Retries in APIs** refer to **repeating a request** to a server after a failure, with the aim of successfully completing the operation. This can occur automatically (by the client or the server) or manually (by the user or system). 

Retries are commonly implemented to **handle transient failures** (temporary issues such as network outages, timeouts, or server overloads) and ensure that operations eventually succeed.

### ✅ **When to Use Retries in APIs:**

1. **Network-related failures:**  
   If there's a network timeout or a transient DNS failure, retrying the request can often solve the issue without requiring human intervention.
   
2. **Server-side failures:**  
   When the server returns a **5xx status code** (e.g., 500 Internal Server Error, 502 Bad Gateway), it often indicates a server problem that may resolve if retried after a short period.

3. **Rate-limiting issues:**  
   If the server returns a `429 Too Many Requests` status, retries can be done with backoff (delaying each retry) to avoid overwhelming the server.

4. **Idempotent operations:**  
   For **idempotent** HTTP methods like `GET`, `PUT`, or `DELETE`, retries are safe since they won’t cause unintended side effects. Retries are typically **safe** for these operations, and you can repeat them without worrying about corrupting data.

### 🚨 **When NOT to Retry:**

1. **Non-idempotent operations (like POST):**  
   For operations like `POST`, which may create resources (e.g., creating a user), retries might result in **duplicate entries**. You should use retries cautiously and only when you are confident that the operation is safe.
   
2. **Client-side errors (4xx responses):**  
   Retries should not be performed for 4xx status codes (like 400 Bad Request, 401 Unauthorized, etc.) since they usually indicate that the client made an error, and retrying won’t help resolve the issue.

---

### ✅ **How to Implement Retries:**

Here’s how retry mechanisms are typically implemented in APIs:

1. **Basic Retry Mechanism:**
   You attempt the request again after a failure, typically with a small delay between attempts.

2. **Exponential Backoff:**
   With **exponential backoff**, you increase the time delay between retries after each failure. This helps reduce the load on the server and gives it time to recover.

   Example:
   - Retry 1: Wait for 1 second
   - Retry 2: Wait for 2 seconds
   - Retry 3: Wait for 4 seconds
   - Retry 4: Wait for 8 seconds
   - And so on...

3. **Max Retries:**
   There’s typically a limit to how many times you retry, to avoid infinite retry loops and to give the server enough time to recover.

---

### 🧰 **Example of Exponential Backoff in Python:**

```python
import time
import random
import requests

def make_request_with_retries(url, max_retries=5):
    retries = 0
    while retries < max_retries:
        try:
            response = requests.get(url)
            response.raise_for_status()  # Raise error for 4xx/5xx status codes
            return response
        except (requests.exceptions.RequestException) as e:
            retries += 1
            wait_time = 2 ** retries + random.random()  # Exponential backoff with jitter
            print(f"Retrying in {wait_time:.2f} seconds...")
            time.sleep(wait_time)
    raise Exception(f"Failed to get response from {url} after {max_retries} retries.")

# Example use case
url = "https://example.com/api"
response = make_request_with_retries(url)
print(response.content)
```

- **`time.sleep(wait_time)`**: Adds a delay before retrying.
- **Exponential backoff**: `2 ** retries` increases the wait time after each failed attempt, with some **random jitter** to avoid thundering herd problems (all clients retrying at the same time).

---

### ✅ **Best Practices for Retries in APIs:**

1. **Limit retries**: Avoid excessive retries. Typically, 3–5 retries are sufficient.
2. **Handle timeouts**: Ensure timeouts are configured appropriately. If the server is taking too long, it's better to give up early than to wait indefinitely.
3. **Log retries**: Keep track of retries, so you know if the request has failed repeatedly.
4. **Exponential backoff**: Implement exponential backoff to reduce the load on both the client and server.
5. **Use status codes**: Only retry on safe status codes, like 5xx errors or 429 (rate-limited). Do not retry on 4xx errors.
6. **Use `idempotent` methods**: Prefer retries for **idempotent** methods like `GET`, `PUT`, and `DELETE` to avoid unintended side effects.

---

### ✅ **Summary:**

- **Retries** in APIs help handle transient failures, allowing operations to succeed despite temporary issues.
- Use **exponential backoff** with a **maximum retry limit** to avoid overwhelming the server and to give it time to recover.
- Be cautious with **non-idempotent operations** (like `POST`), as retries could result in unintended side effects (like duplicate resources).
- Use retries mainly for **idempotent HTTP methods** and safe error conditions like server-side failures (5xx) or rate limiting (429).

### ✅ **1. Timeouts in API Requests:**

**Timeouts** are essential for controlling how long the client waits for a response from the server before giving up. Without proper timeouts, the client could hang indefinitely if the server is taking too long to respond, leading to a poor user experience.

#### Types of Timeouts:

1. **Connection Timeout**:  
   This is the amount of time the client waits while establishing a connection with the server. If the connection cannot be established within the specified time, the request is aborted.
   
2. **Read Timeout**:  
   This is the amount of time the client waits for the server to send back a response after the connection is established. If the server does not respond in this time, the request is aborted.
   
3. **Total Timeout**:  
   Some libraries allow you to set an overall timeout that applies to the entire request, including both connection and reading phases.

#### Example in Python (with `requests` library):

```python
import requests

# Set connection timeout and read timeout
response = requests.get("https://example.com/api", timeout=(5, 10))  # (connection_timeout, read_timeout)
print(response.status_code)
```

- Here, the client will wait up to **5 seconds** for the connection to be established and up to **10 seconds** for the server to send the response after the connection is established.

### ✅ **2. Logging Retries:**

Logging retries is an important practice for monitoring and debugging, especially in production systems. When a request is retried due to failure, it's useful to **log the retry attempt**, including information such as the **failure reason**, **time of retry**, and **current retry attempt number**.

Logging retries helps in:
- Identifying patterns or recurring issues (e.g., repeated server downtime).
- Debugging issues by providing insight into which requests failed and why.
- Monitoring retry behavior and adjusting retry strategies.

#### Example of Logging Retries:

```python
import logging
import time
import random

logging.basicConfig(level=logging.INFO)

def make_request_with_retries(url, max_retries=5):
    retries = 0
    while retries < max_retries:
        try:
            # Simulate request (e.g., using requests.get(url))
            raise Exception("Temporary network issue")  # Simulating failure
        except Exception as e:
            retries += 1
            wait_time = 2 ** retries + random.random()
            logging.warning(f"Attempt {retries} failed: {str(e)}. Retrying in {wait_time:.2f} seconds...")
            time.sleep(wait_time)
    logging.error(f"Failed to get response from {url} after {max_retries} retries.")
```

- **Logging Warning**: Logs retry attempts with reason (`warning` level).
- **Logging Error**: Logs failure after the maximum retries (`error` level).

### ✅ **3. Limiting Retries:**

**Limiting retries** is important to prevent an infinite retry loop that could cause **server overload** or **wasted resources**. Typically, retries should be limited to a small number of attempts (usually between 3 to 5).

#### Why Limit Retries?
- **Resource Conservation**: Constant retries can overload the system, both on the client and server side.
- **Server Health**: If the server is down or struggling, repeatedly calling it could make things worse. Limiting retries gives the server a chance to recover.
- **Efficient Use of Time**: After a certain number of attempts, it’s better to fail fast and inform the user, rather than continuing to attempt and delay the process.

#### Example:

```python
MAX_RETRIES = 5

for attempt in range(1, MAX_RETRIES + 1):
    try:
        # Simulate API call (e.g., requests.get)
        response = "Success"
        break
    except Exception as e:
        if attempt == MAX_RETRIES:
            print(f"Max retries reached. Failing with error: {e}")
        else:
            print(f"Retrying {attempt}/{MAX_RETRIES}...")
```

- This will retry the request up to **5 times**. If it still fails after that, it logs the error and exits.

### ✅ **4. Exponential Backoff:**

**Exponential backoff** is a strategy where the **wait time** between retries increases exponentially after each failure. The idea is to give the server time to recover while preventing an overload of requests.

This is especially useful in cases of transient failures (e.g., server overload, network congestion) to avoid **thundering herd problems**, where multiple clients retry at the same time and make the situation worse.

#### How Exponential Backoff Works:
- On the first retry, the client waits for a small time (e.g., 1 second).
- On the second retry, it waits twice as long (e.g., 2 seconds).
- On the third retry, it waits four times longer (e.g., 4 seconds).
- The backoff continues growing exponentially (doubling the time).

#### Example:

```python
import time
import random

def make_request_with_exponential_backoff(url, max_retries=5):
    retries = 0
    while retries < max_retries:
        try:
            # Simulate request (e.g., using requests.get)
            raise Exception("Temporary network issue")  # Simulating failure
        except Exception as e:
            retries += 1
            wait_time = 2 ** retries + random.random()  # Exponential backoff with random jitter
            print(f"Attempt {retries} failed: {str(e)}. Retrying in {wait_time:.2f} seconds...")
            time.sleep(wait_time)
    print(f"Failed to get response from {url} after {max_retries} retries.")

# Example call
make_request_with_exponential_backoff("https://example.com/api")
```

- The `wait_time = 2 ** retries + random.random()` line gives **exponential backoff** combined with **random jitter** (using `random.random()` to add randomness and avoid all clients retrying at the same exact time).

---

### ✅ **Summary of Concepts:**

| Concept                | Explanation                                                                 |
|------------------------|-----------------------------------------------------------------------------|
| **Timeouts**            | Specifies how long to wait for a response (Connection Timeout and Read Timeout). Prevents hanging indefinitely. |
| **Logging Retries**     | Logs retry attempts, including failure reasons, to help with debugging and monitoring. |
| **Limiting Retries**    | Restricts the number of retries to prevent overloading the server and wasting resources. |
| **Exponential Backoff** | Increases the wait time between retries, typically doubling the delay after each failed attempt, to prevent thundering herd problems. |

### **What is a Mutex?**

A **mutex** (short for "mutual exclusion") is a synchronization primitive that is used to **prevent multiple threads from accessing a shared resource concurrently** in a multithreaded program. It ensures that only one thread at a time can acquire the lock and access the shared resource, providing **exclusive access**.

When a thread locks a mutex, other threads are blocked from accessing the locked resource until the mutex is unlocked by the thread that holds the lock. 

A mutex is typically used for situations where:
- You have shared resources that must not be accessed simultaneously (e.g., variables, files, databases).
- You need to enforce **exclusive access** to a resource across threads or processes.

#### Key Characteristics of a Mutex:
- **Binary Lock**: A mutex can either be **locked** (by a thread) or **unlocked**.
- **Ownership**: Only the thread that locks a mutex can unlock it. This is a key distinction from semaphores, where any thread can signal the semaphore.
- **Blocking**: When a thread tries to acquire a locked mutex, it will **block** (i.e., wait) until the mutex becomes available.

### **What is a Semaphore?**

A **semaphore** is a more general synchronization primitive compared to a mutex. It is used to control access to a **shared resource pool** that has a limited number of resources. Semaphores allow multiple threads to acquire and release access to resources, with the semaphore keeping track of how many resources are available.

There are two types of semaphores:
1. **Counting Semaphore**: Allows a thread to acquire a resource if the semaphore's counter is greater than 0. The counter is incremented when a resource is released and decremented when a resource is acquired.
2. **Binary Semaphore**: Similar to a mutex, it only has two states (locked/unlocked), and it can be used as a signaling mechanism. A **binary semaphore** is sometimes referred to as a **flag**.

#### Key Characteristics of a Semaphore:
- **Count-Based**: The semaphore maintains a count, and threads can access the shared resource as long as the count is greater than 0.
- **No Ownership**: Unlike mutexes, semaphores do not have ownership. Any thread can signal (release) the semaphore or acquire (wait) the resource.
- **Non-Blocking**: If no resources are available (the count is 0), a thread will block, but there is no concept of thread ownership like with a mutex.

### **Mutex vs. Semaphore**

| Feature             | **Mutex**                                  | **Semaphore**                           |
|---------------------|--------------------------------------------|-----------------------------------------|
| **Purpose**          | Used to **lock** shared resources, ensuring exclusive access. | Used to **control access** to a pool of resources (can allow multiple threads). |
| **State**            | Binary (locked/unlocked).                  | Counting or binary (count can be greater than 1 for resource pools). |
| **Thread Ownership** | Only the thread that locks the mutex can unlock it. | Any thread can signal (release) or wait (acquire) on the semaphore. |
| **Blocking**         | Threads **block** until the mutex is unlocked by the owning thread. | Threads **block** when the semaphore count is 0, but not based on ownership. |
| **Use Case**         | Enforces **exclusive** access to a single resource. | Controls access to a **limited number of resources** (e.g., a fixed pool). |
| **Deadlock Risk**    | Deadlock can occur if multiple threads try to lock mutexes in different orders. | Less prone to deadlocks (unless misused with incorrect counting). |
| **Examples**         | Ensuring that only one thread can update a global variable. | Managing access to a fixed number of database connections. |

### **Key Differences:**
1. **Mutex**:
   - **Exclusive lock**: Only one thread can hold the mutex at a time.
   - **Ownership**: Mutex is owned by the thread that locks it, and only that thread can unlock it.
   - Used to protect a **single shared resource**.

2. **Semaphore**:
   - **Resource pool**: Semaphore controls access to multiple instances of a resource, allowing multiple threads to acquire access concurrently (up to the semaphore count).
   - **No ownership**: Any thread can signal or wait on a semaphore.
   - Used for managing access to a **pool of resources** (e.g., limiting the number of concurrent database connections).

---

### **Example Use Cases:**

1. **Mutex Example (Exclusive Access)**:
   A shared **printer** resource should only be accessed by one thread at a time.

   ```python
   import threading

   # Creating the mutex
   printer_mutex = threading.Lock()

   def print_document():
       with printer_mutex:  # Only one thread can enter this block at a time
           print("Printing document...")

   # Simulating multiple threads
   thread1 = threading.Thread(target=print_document)
   thread2 = threading.Thread(target=print_document)

   thread1.start()
   thread2.start()

   thread1.join()
   thread2.join()
   ```

2. **Semaphore Example (Resource Pool)**:
   A shared **connection pool** with a limited number of connections can be managed using a semaphore. Suppose the server only allows 3 concurrent connections at a time.

   ```python
   import threading
   import time

   # Semaphore with 3 resources (connections)
   semaphore = threading.Semaphore(3)

   def access_resource(thread_name):
       with semaphore:  # Acquire a connection
           print(f"{thread_name} accessing the resource")
           time.sleep(2)  # Simulate some work
           print(f"{thread_name} releasing the resource")

   # Simulate 5 threads trying to access the resource
   threads = [threading.Thread(target=access_resource, args=(f"Thread-{i}",)) for i in range(5)]

   for t in threads:
       t.start()

   for t in threads:
       t.join()
   ```

In this example, only 3 threads can access the shared resource concurrently, and the rest will have to wait until a connection is released.

---

### **Summary:**

- **Mutex** is used for **exclusive locking** of a resource by one thread at a time, and it has **ownership**.
- **Semaphore** is used to manage access to a **pool of resources** and does not have ownership, meaning any thread can acquire or release a semaphore.

### **Q: What are SOLID Principles?**

**Answer**:  
SOLID is an acronym representing five design principles intended to make software designs **more understandable, flexible, and maintainable**. These principles help us **write clean, loosely coupled, and scalable code**, especially in object-oriented programming.

---

### 🔹 **S - Single Responsibility Principle (SRP)**  
**Definition**: A class should have only **one reason to change**—it should have only one responsibility.

**In practice**: Each class/module/function should do **one thing well**.

**Example**:
```python
# ✅ Good SRP
class Invoice:
    def calculate_total(self):
        # logic

class InvoicePrinter:
    def print_invoice(self, invoice):
        # logic

class InvoiceSaver:
    def save_to_db(self, invoice):
        # logic
```

📌 **Why?** Changes in print/save logic won’t affect calculation logic.

---

### 🔹 **O - Open/Closed Principle (OCP)**  
**Definition**: Software entities should be **open for extension** but **closed for modification**.

**In practice**: You should be able to add new functionality **without modifying existing code**.

**Example (Strategy Pattern)**:
```python
class DiscountStrategy:
    def apply(self, amount):
        pass

class PercentageDiscount(DiscountStrategy):
    def apply(self, amount):
        return amount * 0.9

class FlatDiscount(DiscountStrategy):
    def apply(self, amount):
        return amount - 50
```

📌 New strategies can be added without changing existing logic.

---

### 🔹 **L - Liskov Substitution Principle (LSP)**  
**Definition**: Objects of a superclass should be **replaceable with objects of its subclasses** without breaking the application.

**In practice**: Subclasses should **honor the contract** of the parent class.

**Example**:
```python
class Bird:
    def fly(self): pass

class Sparrow(Bird):
    def fly(self): pass

class Ostrich(Bird):
    def fly(self): raise NotImplementedError("Ostriches can't fly!")  # ❌ violates LSP
```

📌 Solution: Split the hierarchy — don't force subclasses to inherit methods they can't implement.

---

### 🔹 **I - Interface Segregation Principle (ISP)**  
**Definition**: Clients should not be forced to depend on **interfaces they do not use**.

**In practice**: Break large interfaces into **smaller, more specific ones**.

**Example**:
```python
# ❌ Bad interface
class MultiFunctionDevice:
    def print(self): pass
    def scan(self): pass
    def fax(self): pass

# ✅ ISP applied
class Printer:
    def print(self): pass

class Scanner:
    def scan(self): pass

class Fax:
    def fax(self): pass
```

📌 A device can implement only what it needs.

---

### 🔹 **D - Dependency Inversion Principle (DIP)**  
**Definition**: High-level modules should not depend on low-level modules. **Both should depend on abstractions**.

**In practice**: Rely on interfaces/abstract classes rather than concrete implementations.

**Example**:
```python
# Abstraction
class MessageSender:
    def send(self, message): pass

# Low-level implementation
class EmailSender(MessageSender):
    def send(self, message):
        print(f"Email sent: {message}")

# High-level module
class NotificationService:
    def __init__(self, sender: MessageSender):
        self.sender = sender

    def notify(self, msg):
        self.sender.send(msg)
```

📌 You can inject any `MessageSender` (Email, SMS, Push) without changing the `NotificationService`.

---

### ✅ Summary:

| Principle | Goal |
|----------|------|
| **S** | A class does one thing |
| **O** | Add features without changing existing code |
| **L** | Subtypes behave like their parents |
| **I** | No forced implementation of unused methods |
| **D** | Depend on abstractions, not concretes |

### **Q: What are Green Threads?**

**Answer**:  
**Green threads** are **user-level threads** that are **scheduled and managed by a runtime library or virtual machine (VM)** instead of the operating system. They're called *"green"* threads because they don’t rely on native OS-level threading—they’re implemented in “user space.”

---

### ✅ **Key Characteristics of Green Threads**:
- **Not managed by OS**: They are **simulated** by the programming language runtime (like the JVM or Python interpreter).
- **Single OS thread**: Multiple green threads can run on **a single OS thread**.
- **Cooperative multitasking**: Green threads often rely on **cooperative scheduling**—the thread itself yields control, rather than being preempted by the OS.
- **No true parallelism**: Since they're on one OS thread, they **can’t take advantage of multiple CPU cores**.

---

### 🧠 **Why use Green Threads?**
- **Lightweight**: Creating green threads is faster and consumes less memory than native threads.
- **Portable**: Easier to implement in environments where native threads aren't fully supported (early JVMs, embedded systems).
- **Efficient for I/O-bound tasks**: They're great for handling many concurrent I/O operations because they can yield when waiting.

---

### 🔁 **Green Threads vs Native (OS) Threads**

| Feature | **Green Threads** | **OS Threads (Kernel Threads)** |
|--------|-------------------|-------------------------------|
| Scheduling | Done by **language runtime** | Done by **OS scheduler** |
| Parallelism | **No true parallelism** | Can run **on multiple cores** |
| Performance | Lightweight, but limited by one core | Heavier, but scalable |
| Blocking calls | Can **block all green threads** if one blocks | Blocking one doesn’t block others |
| Use case | I/O-bound, high concurrency | CPU-bound, parallel tasks |

---

### 🧪 **Examples**:

#### ☕ Java (before JDK 1.2):
Older Java VMs used green threads—so even if a system had multiple CPUs, Java threads wouldn’t run in parallel.

#### 🐍 Python (with asyncio):
Python’s `asyncio` is conceptually similar—it runs many lightweight tasks cooperatively using a single OS thread. These are **not literally green threads**, but follow the same philosophy: **user-managed concurrency** with no parallelism.

---

### 🧰 Green Threads in Practice:

- **Languages/Frameworks** that support green-like threading:
  - **Golang**: Goroutines are **user-space threads** scheduled by the Go runtime (not OS-level threads).
  - **Erlang**: Uses lightweight processes that behave like green threads.
  - **Python `asyncio`**: Cooperative multitasking with async/await (not technically green threads, but follows similar principles).
  - **Java Virtual Machine** (early versions).
  - **Kotlin Coroutines**: Also built on suspend/resume and cooperative concurrency.

---

### ✅ Summary:
> Green threads are **user-level threads** managed by a **runtime** rather than the OS. They’re efficient for high-concurrency, I/O-bound applications but **lack parallelism** because they share one OS thread.

### **Q: How is Inter-Process Communication (IPC) or Inter-Thread Communication handled in Python?**

**Answer**:  
In Python, **IPC depends on whether you're using threads or processes**, because threads share memory, but processes do not. So the way you communicate and share data differs.

---

## 🧵 **Threads: Communication is simpler**
Since **threads share the same memory space**, communication is often done using **shared objects and synchronization mechanisms**.

### ✅ Communication between Threads:
You can use:
- **Shared variables** (since they live in the same address space)
- **Queues** (`queue.Queue`) – *thread-safe FIFO structure*
- **Locks / Semaphores / Events / Conditions** (from `threading` module) – for synchronization, not data passing

```python
from queue import Queue
import threading

q = Queue()

def producer():
    for i in range(5):
        q.put(i)

def consumer():
    while not q.empty():
        print(q.get())

t1 = threading.Thread(target=producer)
t2 = threading.Thread(target=consumer)
t1.start()
t2.start()
```

> 🧠 Threads **do not need special IPC tools** because they share memory.

---

## 🔀 **Processes: Need explicit IPC**

Since **processes have separate memory spaces**, Python provides the `multiprocessing` module which offers multiple IPC tools.

---

### ✅ Communication between **Processes**:

#### 1. **Queue** (`multiprocessing.Queue`)
- Just like thread-safe `Queue`, but backed by a pipe.
- Best for passing Python objects between processes.

```python
from multiprocessing import Process, Queue

def f(q):
    q.put([1, 2, 3])

q = Queue()
p = Process(target=f, args=(q,))
p.start()
print(q.get())  # Output: [1, 2, 3]
p.join()
```

---

#### 2. **Pipe** (`multiprocessing.Pipe`)
- Two-way communication channel.
- More low-level than `Queue`.

```python
from multiprocessing import Process, Pipe

def f(conn):
    conn.send('Hello')
    conn.close()

parent_conn, child_conn = Pipe()
p = Process(target=f, args=(child_conn,))
p.start()
print(parent_conn.recv())  # Output: Hello
p.join()
```

---

#### 3. **Value / Array** (`multiprocessing.Value`, `multiprocessing.Array`)
- For sharing a **single value or array** between processes (in shared memory).
- Useful for numeric data that needs to be updated frequently.

```python
from multiprocessing import Process, Value

def f(n):
    n.value += 1

num = Value('i', 5)  # 'i' = int
p = Process(target=f, args=(num,))
p.start()
p.join()
print(num.value)  # Output: 6
```

---

#### 4. **Manager** (`multiprocessing.Manager`)
- Allows sharing **complex Python objects** like dicts, lists, namespaces between processes.

```python
from multiprocessing import Process, Manager

def f(d):
    d['key'] = 'value'

with Manager() as manager:
    shared_dict = manager.dict()
    p = Process(target=f, args=(shared_dict,))
    p.start()
    p.join()
    print(shared_dict)  # Output: {'key': 'value'}
```

---

### 🧠 Summary Table:

| Communication Tool     | Threads | Processes | Use Case |
|------------------------|---------|-----------|----------|
| `queue.Queue`          | ✅      | ❌        | Thread-safe data passing |
| `multiprocessing.Queue`| ❌      | ✅        | Process-safe queue |
| `multiprocessing.Pipe` | ❌      | ✅        | Two-way channel |
| `Value` / `Array`      | ❌      | ✅        | Shared memory (simple data) |
| `Manager().dict()`     | ❌      | ✅        | Share complex objects |
| Shared variables       | ✅      | ❌        | Easy but needs locking |
| `threading.Lock` etc.  | ✅      | ❌        | For thread sync |

---

### ✅ TL;DR:
- **Threads**: Share memory → communicate via shared variables, `queue.Queue`, and sync primitives.
- **Processes**: Don’t share memory → use `Queue`, `Pipe`, `Value`, `Array`, or `Manager` for communication.

### **Q: What is a Deadlock, Race Condition, and how does Synchronization help?**

---

### 🔐 **1. What is a Race Condition?**

**Answer**:  
A **race condition** occurs when **two or more threads or processes access shared data at the same time**, and **the outcome depends on the order** in which the access happens. If not properly synchronized, this can lead to **unpredictable or incorrect behavior**.

> 🧠 It’s called a “race” because the threads are *racing* to access/modify the same resource.

---

#### ✅ Example:
```python
# Two threads incrementing a shared counter

counter = 0

def increment():
    global counter
    for _ in range(100000):
        counter += 1

t1 = threading.Thread(target=increment)
t2 = threading.Thread(target=increment)
t1.start()
t2.start()
t1.join()
t2.join()
print(counter)  # Expected: 200000, Actual: unpredictable!
```

> Here, both threads read and write to `counter` at the same time, leading to a **race condition**.

---

### 🧱 **2. What is a Deadlock?**

**Answer**:  
A **deadlock** happens when **two or more threads are waiting indefinitely for resources** that are held by each other, and **none of them can proceed**.

#### ✅ Classic Scenario:
- Thread A holds **Lock 1** and waits for **Lock 2**.
- Thread B holds **Lock 2** and waits for **Lock 1**.
- Both are waiting for each other → **Deadlock**.

---

#### ⚠️ Example:
```python
lock1 = threading.Lock()
lock2 = threading.Lock()

def thread1():
    with lock1:
        print("Thread 1 acquired lock1")
        with lock2:
            print("Thread 1 acquired lock2")

def thread2():
    with lock2:
        print("Thread 2 acquired lock2")
        with lock1:
            print("Thread 2 acquired lock1")

t1 = threading.Thread(target=thread1)
t2 = threading.Thread(target=thread2)
t1.start()
t2.start()
t1.join()
t2.join()
```

> This may result in a **deadlock** depending on timing.

---

### 🔧 **3. What is Synchronization?**

**Answer**:  
**Synchronization** is a way to **control access to shared resources** to avoid race conditions and deadlocks. In Python, you can use primitives from the `threading` or `multiprocessing` module:

#### ✅ Common Synchronization Tools:
| Tool             | Purpose |
|------------------|---------|
| `Lock`           | Only one thread at a time can access the resource |
| `RLock`          | Reentrant lock (same thread can acquire multiple times) |
| `Semaphore`      | Allows limited concurrent access |
| `Event`          | Used for signaling between threads |
| `Condition`      | Complex coordination (like producer-consumer) |

---

#### 🧪 Synchronizing our earlier race condition:

```python
lock = threading.Lock()
counter = 0

def increment():
    global counter
    for _ in range(100000):
        with lock:
            counter += 1

# Now safe from race conditions
```

---

### 🧠 Summary:

| Concept         | Definition |
|----------------|------------|
| **Race Condition** | Multiple threads access shared data without proper sync → unpredictable results |
| **Deadlock** | Threads are waiting on each other indefinitely, blocking progress |
| **Synchronization** | Mechanism to safely coordinate access to shared resources |

---

> ✅ "Synchronization prevents race conditions. Avoiding deadlocks requires careful design."


## 🧰 1. **Singleton Pattern**

**Concept**:  
Ensure only **one instance** of a class is created, and provide a global point of access to it.

### ✅ Use Case:
- Database connections
- Logger
- Config manager

### 🧑‍💻 Python Code:
```python
import threading

class SingletonMeta(type):
    _instances = {}
    _lock = threading.Lock()  # Thread-safe Singleton

    def __call__(cls, *args, **kwargs):
        with cls._lock:
            if cls not in cls._instances:
                cls._instances[cls] = super().__call__(*args, **kwargs)
        return cls._instances[cls]

class Logger(metaclass=SingletonMeta):
    def log(self, message):
        print(f"[LOG]: {message}")
```

### ✅ Thread Safe?  
Yes – if implemented with a lock like above.

---

## 🏭 2. **Factory Pattern**

**Concept**:  
Creates objects **without exposing instantiation logic** to the client. It lets the subclass decide which class to instantiate.

### ✅ Use Case:
- UI components (`Button` → MacButton, WindowsButton)
- Notification systems (Email, SMS, Push)

### 🧑‍💻 Python Code:
```python
class Notification:
    def notify(self):
        raise NotImplementedError

class EmailNotification(Notification):
    def notify(self):
        print("Sending Email")

class SMSNotification(Notification):
    def notify(self):
        print("Sending SMS")

class NotificationFactory:
    def create_notification(self, channel):
        if channel == "email":
            return EmailNotification()
        elif channel == "sms":
            return SMSNotification()
        else:
            raise ValueError("Unknown channel")

# Client
factory = NotificationFactory()
notifier = factory.create_notification("email")
notifier.notify()
```

### ✅ Thread Safe?  
Not inherently needed unless the factory itself holds shared state.

---

## 🧱 3. **Abstract Factory Pattern**

**Concept**:  
Creates families of related objects without specifying their concrete classes.

### ✅ Use Case:
- UI themes (LightFactory, DarkFactory → Button, Dropdown)
- Cross-platform UI toolkits

### 🧑‍💻 Python Code:
```python
class Button:
    def render(self): pass

class LightButton(Button):
    def render(self): print("Light Button")

class DarkButton(Button):
    def render(self): print("Dark Button")

class UIFactory:
    def create_button(self): pass

class LightFactory(UIFactory):
    def create_button(self): return LightButton()

class DarkFactory(UIFactory):
    def create_button(self): return DarkButton()

# Client
factory = DarkFactory()
button = factory.create_button()
button.render()  # Output: Dark Button
```

### ✅ Thread Safe?  
Only if factories are shared and mutable – otherwise usually safe.

---

## 🧱 4. **Builder Pattern**

**Concept**:  
Separates the construction of a complex object from its representation.

### ✅ Use Case:
- Building complex `HTML`, `SQL`, or `PDF` objects step by step
- Fluent interfaces (like in test cases)

### 🧑‍💻 Python Code:
```python
class Burger:
    def __init__(self):
        self.ingredients = []

    def __str__(self):
        return "Burger with " + ", ".join(self.ingredients)

class BurgerBuilder:
    def __init__(self):
        self.burger = Burger()

    def add_patty(self): self.burger.ingredients.append("Patty"); return self
    def add_lettuce(self): self.burger.ingredients.append("Lettuce"); return self
    def add_cheese(self): self.burger.ingredients.append("Cheese"); return self
    def build(self): return self.burger

# Client
burger = BurgerBuilder().add_patty().add_cheese().build()
print(burger)  # Burger with Patty, Cheese
```

### ✅ Thread Safe?  
Only if builders are reused across threads (avoid reusing the same builder instance).

---

## 🔁 5. **Prototype Pattern**

**Concept**:  
Create new objects by copying an existing object (clone), instead of instantiating new ones.

### ✅ Use Case:
- Game objects, when creating enemies with variations
- Large object copies (deepcopy performance optimization)

### 🧑‍💻 Python Code:
```python
import copy

class Shape:
    def __init__(self, color):
        self.color = color

    def clone(self):
        return copy.deepcopy(self)

circle = Shape("Red")
another_circle = circle.clone()
print(another_circle.color)  # Red
```

### ✅ Thread Safe?  
Yes, if cloned objects are used independently.

---

## 🔌 6. **Adapter Pattern**

**Concept**:  
Convert the interface of one class into another that the client expects. It’s like a wrapper.

### ✅ Use Case:
- Integrating old APIs into a new system
- Adapting interfaces in third-party libraries

### 🧑‍💻 Python Code:
```python
class OldPrinter:
    def old_print(self): print("Old Printer")

class NewPrinter:
    def new_print(self): print("New Printer")

class PrinterAdapter:
    def __init__(self, obj):
        self.obj = obj

    def print(self):
        if hasattr(self.obj, 'new_print'):
            self.obj.new_print()
        else:
            self.obj.old_print()

# Client
printer = PrinterAdapter(NewPrinter())
printer.print()  # Output: New Printer
```

### ✅ Thread Safe?  
As long as adapter instances aren’t shared/mutated across threads, yes.

## 📢 **Observer / Publish-Subscribe Pattern**

---

### ✅ **Concept**

> The **Observer Pattern** defines a **one-to-many dependency** between objects. When one object (the **Subject** or **Publisher**) changes its state, **all its dependents (Observers/Subscribers)** are notified automatically.

This pattern **decouples the sender from the receivers**, meaning the subject doesn’t need to know who or how many observers exist.

---

| Use Case                           | Description                                              |
|------------------------------------|----------------------------------------------------------|
| UI Frameworks                      | Button click → notify event listeners                    |
| Messaging Systems (e.g., Kafka)    | Producer sends message → multiple consumers process it   |
| Stock Ticker Apps                  | Stock price changes → all registered apps get updates    |
| Social Media Notifications         | You follow someone → you get notified on their activity  |
| Event-driven architectures         | Microservices emitting domain events                     |

## 🧑‍💻 Python Code Example

### 🧠 Simple Observer Pattern:
```python
# Subject (Publisher)
class NewsPublisher:
    def __init__(self):
        self.subscribers = []

    def subscribe(self, subscriber):
        self.subscribers.append(subscriber)

    def unsubscribe(self, subscriber):
        self.subscribers.remove(subscriber)

    def notify(self, news):
        for subscriber in self.subscribers:
            subscriber.update(news)

# Observer (Subscriber)
class EmailSubscriber:
    def update(self, news):
        print(f"Email received: {news}")

class SMSSubscriber:
    def update(self, news):
        print(f"SMS received: {news}")

# Client Code
publisher = NewsPublisher()
email_sub = EmailSubscriber()
sms_sub = SMSSubscriber()

publisher.subscribe(email_sub)
publisher.subscribe(sms_sub)

publisher.notify("Breaking News: Observer pattern explained!")
```

🟢 Output:
```
Email received: Breaking News: Observer pattern explained!
SMS received: Breaking News: Observer pattern explained!
```

## 🔁 **Push vs Pull Models**

| Model | Explanation | Example |
|-------|-------------|---------|
| **Push** | Publisher sends the full data to subscribers | `subscriber.update(data)` |
| **Pull** | Publisher just notifies, and subscribers fetch what they need | `subscriber.update()` → internally fetches data |

> Push is simple and direct. Pull is more flexible and reduces data sent unnecessarily.

## 🧵 **Is Observer Thread Safe?**

Not inherently.

- If multiple threads subscribe/unsubscribe or receive notifications, use thread-safe collections and locks.

```python
import threading

class ThreadSafePublisher:
    def __init__(self):
        self._subscribers = []
        self._lock = threading.Lock()

    def subscribe(self, subscriber):
        with self._lock:
            self._subscribers.append(subscriber)

    def notify(self, news):
        with self._lock:
            for sub in self._subscribers:
                sub.update(news)
```

## ⚙️ What is Event-Driven Architecture?

> Event-Driven Architecture (EDA) is a **software design pattern** in which **events** (changes in state) drive the **communication between decoupled components**.

In this model, **producers emit events**, and **consumers react to them**, often asynchronously.

---

## 🧩 Key Components

| Component        | Role                                                                 |
|------------------|----------------------------------------------------------------------|
| **Event**        | A message or signal that something has happened (e.g. "order placed")|
| **Producer**     | Component that **emits** events                                       |
| **Consumer**     | Component that **reacts** to events                                   |
| **Event Broker** | Middleware that routes events (e.g. Kafka, RabbitMQ, AWS SNS/SQS)     |

---

## 🧠 Interview-Friendly Analogy

> Think of an **event** like a notification in a restaurant:  
When an order is ready, the kitchen (producer) rings a bell (event), and the waiter (consumer) comes to pick it up.

---

## 🔄 Flow in Event-Driven Systems

```plaintext
Component A  ---> [ emits event: "User Signed Up" ] ---> Event Broker ---> Component B (sends welcome email)
                                                                   ---> Component C (logs to analytics)
                                                                   ---> Component D (creates default user profile)
```

Each of the consumers independently acts on the event — decoupled from the producer.

---

## ✅ Benefits

| Benefit               | Why it matters                                                  |
|------------------------|----------------------------------------------------------------|
| **Loose coupling**     | Producers and consumers don’t need to know about each other    |
| **Scalability**        | Services can scale independently                               |
| **Asynchronous flow**  | Improves performance and responsiveness                        |
| **Extensibility**      | New consumers can be added without changing the producer       |
| **Fault Isolation**    | A failure in one consumer doesn’t impact others                |

## 🧑‍💻 Python Example (Minimal In-Memory)

```python
from collections import defaultdict

class EventBus:
    def __init__(self):
        self.subscribers = defaultdict(list)

    def subscribe(self, event_type, handler):
        self.subscribers[event_type].append(handler)

    def publish(self, event_type, data):
        for handler in self.subscribers[event_type]:
            handler(data)

# Define some event handlers (consumers)
def send_welcome_email(user):
    print(f"Sending welcome email to {user}")

def log_signup(user):
    print(f"Logging signup for {user}")

# Set up the event bus
bus = EventBus()
bus.subscribe("user_signed_up", send_welcome_email)
bus.subscribe("user_signed_up", log_signup)

# Emit event
bus.publish("user_signed_up", "pratyusha@example.com")
```

🟢 Output:
```
Sending welcome email to pratyusha@example.com
Logging signup for pratyusha@example.com
```

## 🧵 Event-Driven vs Request-Driven (REST)

| Aspect               | Event-Driven                          | Request-Driven (REST)               |
|----------------------|----------------------------------------|-------------------------------------|
| Communication        | Asynchronous                          | Synchronous                         |
| Decoupling           | High (no direct dependency)           | Tight coupling                      |
| Reliability Tools    | Message queues, retry mechanisms      | HTTP retries                        |
| Use Cases            | Real-time, background processing      | CRUD operations                     |
